# 1 Question

What is an API? Why are APIs important?


## Answer

## What is an API?

**API (Application Programming Interface)** is a set of rules that allows different software applications to communicate with each other.

It acts like a bridge between two systems.

---

## Example

When you use a weather app:

- The app sends a request to a weather server using an API
- The server sends weather data back to the app

---

## Why APIs are Important

### 1. Communication Between Systems

APIs allow applications to exchange data and functionality.

Example:

- Payment gateway integration
- Google Maps integration

---

### 2. Saves Development Time

Developers can use existing services instead of building everything from scratch.

---

### 3. Improves Scalability

Different services can work independently and communicate using APIs.

---

### 4. Enables Integration

APIs help connect:

- Mobile apps
- Websites
- Databases
- Cloud services

---

## Real-Life Examples

| Application | API Used |
|---|---|
| Food Delivery App | Payment API |
| Social Media Login | Google/Facebook API |
| E-commerce Website | Shipping API |

---

## Explanation

APIs are essential because they enable communication and integration between different software systems efficiently.

# 2 Question

Explain the core principles of RESTful APIs.


## Answer

## What is REST?

**REST (Representational State Transfer)** is an architectural style used for designing web APIs.

REST APIs use HTTP methods to perform operations on resources.

---

## Core Principles of RESTful APIs

### 1. Client-Server Architecture

The client and server are separate.

- Client → Sends requests
- Server → Sends responses

---

### 2. Statelessness

Each request contains all necessary information.

The server does not remember previous requests.

---

### 3. Resource-Based

Everything is treated as a resource.

Examples:

```text
/products
/users
/orders
```

---

### 4. Uniform Interface

REST APIs use standard HTTP methods:

- GET
- POST
- PUT
- DELETE

---

### 5. Representation of Resources

Data is usually transferred in JSON format.

Example:

```json
{
  "id": 1,
  "name": "Laptop"
}
```

---

### 6. Layered System

The client does not know whether it is connected directly to the server or through intermediaries.

---

## Explanation

REST principles make APIs scalable, simple, and easy to maintain.

# 3 Question

What are the common HTTP methods used in REST APIs and what are their typical uses?


## Answer

## Common HTTP Methods

| Method | Purpose | Example |
|---|---|---|
| **GET** | Retrieve data | Fetch products |
| **POST** | Create new data | Add new product |
| **PUT** | Update entire resource | Update product |
| **PATCH** | Update partial resource | Change product price |
| **DELETE** | Remove resource | Delete product |

---

## Examples

### GET

Fetch all users:

```http
GET /users
```

---

### POST

Create a new user:

```http
POST /users
```

---

### PUT

Update user completely:

```http
PUT /users/1
```

---

### PATCH

Update only email:

```http
PATCH /users/1
```

---

### DELETE

Delete a user:

```http
DELETE /users/1
```

---

## Explanation

HTTP methods define the type of action performed on resources in REST APIs.

# 4 Question

Explain the difference between stateless and stateful REST APIs.


## Answer

## Stateless API

A stateless API does not store client information between requests.

Each request must contain all required data.

---

## Example

```http
GET /profile
Authorization: Bearer token123
```

The server processes the request independently.

---

## Advantages

- Scalable
- Faster
- Easier to maintain

---

# Stateful API

A stateful API stores client session information on the server.

The server remembers previous interactions.

---

## Example

Login session stored on server:

```text
User logs in → Server stores session
```

Next requests use the stored session.

---

## Advantages

- Better for long user sessions
- Easier session management

---

## Comparison Table

| Feature | Stateless API | Stateful API |
|---|---|---|
| Server Memory | Does not store session | Stores session |
| Scalability | High | Lower |
| Performance | Faster | Slower |
| Maintenance | Easier | More complex |

---

## Explanation

REST APIs are generally stateless because stateless systems scale better and are simpler.

# 5 Question

Given an e-commerce application, design the URLs for the following REST API endpoints:

- Fetch details of a specific product with ID 123
- Fetch details of all products
- Create a new product
- Delete a specific product with ID 123
- Fetch all reviews for a specific product with ID 123


## Answer

## REST API Endpoints

### 1. Fetch details of a specific product with ID 123

```http
GET /products/123
```

---

### 2. Fetch details of all products

```http
GET /products
```

---

### 3. Create a new product

```http
POST /products
```

Example Request Body:

```json
{
  "name": "Laptop",
  "price": 50000,
  "brand": "Dell"
}
```

---

### 4. Delete a specific product with ID 123

```http
DELETE /products/123
```

---

### 5. Fetch all reviews for a specific product with ID 123

```http
GET /products/123/reviews
```

---

## Explanation

REST APIs use resource-based URLs and HTTP methods to perform operations clearly and efficiently.
