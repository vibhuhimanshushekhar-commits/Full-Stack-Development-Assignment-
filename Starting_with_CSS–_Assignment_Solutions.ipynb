# 1. Create a simple page with some div tags and show different ways to add CSS.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Ways to Add CSS</title>

    <!-- Internal CSS -->
    <style>

        /* Styling all divs using internal CSS */

        div {
            padding: 20px;
            margin: 10px;
            border: 2px solid black;
        }

        .internal-style {
            background-color: lightblue;
            color: darkblue;
        }

    </style>

    <!-- External CSS -->
    <link rel="stylesheet" href="style.css">

</head>
<body>

    <!-- Inline CSS -->
    <div style="background-color: lightgreen; color: darkgreen;">
        This div uses Inline CSS.
    </div>

    <!-- Internal CSS -->
    <div class="internal-style">
        This div uses Internal CSS.
    </div>

    <!-- External CSS -->
    <div class="external-style">
        This div uses External CSS.
    </div>

</body>
</html>
```

### External CSS File (style.css)

```css
/* External CSS */

.external-style {
    background-color: lightcoral;
    color: white;
    font-weight: bold;
}
```

### Explanation:

- Inline CSS has the highest priority.
- Internal CSS is written inside the `<style>` tag.
- External CSS is written in a separate `.css` file.

# 2. Build an HTML page with multiple paragraphs using class selectors and BEM naming convention.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>BEM Naming Example</title>

    <style>

        .content__paragraph--primary {
            color: blue;
            font-size: 18px;
        }

        .content__paragraph--secondary {
            color: green;
            font-size: 16px;
        }

        .content__paragraph--highlight {
            color: red;
            background-color: yellow;
            font-weight: bold;
        }

    </style>

</head>
<body>

    <h1>BEM Naming Convention</h1>

    <p class="content__paragraph--primary">
        This is the primary paragraph.
    </p>

    <p class="content__paragraph--secondary">
        This is the secondary paragraph.
    </p>

    <p class="content__paragraph--highlight">
        This paragraph is highlighted.
    </p>

</body>
</html>
```

### Explanation of BEM Naming:

- `content` → Block
- `paragraph` → Element
- `primary`, `secondary`, `highlight` → Modifiers

BEM improves code readability and maintainability.

# 3. Develop an HTML form with various input elements and style it using CSS.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Styled Form</title>

    <style>

        body {
            font-family: Arial, sans-serif;
            background-color: #f2f2f2;
        }

        .form-container {

            width: 400px;
            margin: 50px auto;
            padding: 20px;

            background-color: white;

            border-radius: 10px;

            box-shadow: 0px 0px 10px gray;
        }

        h2 {
            text-align: center;
            color: darkblue;
        }

        input,
        select,
        textarea {

            width: 100%;
            padding: 10px;
            margin-top: 10px;
            margin-bottom: 15px;

            border: 1px solid gray;
            border-radius: 5px;

            background-color: #f9f9f9;
        }

        .opacity-section {

            background-color: lightblue;
            padding: 15px;

            opacity: 0.7;

            border-radius: 5px;
        }

        button {

            width: 100%;
            padding: 10px;

            background-color: darkblue;
            color: white;

            border: none;
            border-radius: 5px;

            cursor: pointer;
        }

        button:hover {
            background-color: navy;
        }

    </style>

</head>
<body>

    <div class="form-container">

        <h2>Registration Form</h2>

        <form>

            <label>Full Name:</label>
            <input type="text" placeholder="Enter your name">

            <label>Email:</label>
            <input type="email" placeholder="Enter your email">

            <label>Password:</label>
            <input type="password" placeholder="Enter password">

            <label>Gender:</label>

            <select>
                <option>Male</option>
                <option>Female</option>
                <option>Other</option>
            </select>

            <div class="opacity-section">

                <label>Address:</label>

                <textarea rows="4" placeholder="Enter your address"></textarea>

            </div>

            <button type="submit">
                Submit
            </button>

        </form>

    </div>

</body>
</html>
```
