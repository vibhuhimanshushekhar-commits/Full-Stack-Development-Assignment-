# Question 1: Explain the basic structure of an HTML page. Why is it important to follow **this** structure?

## Answer:

The basic structure of an HTML page consists of a few essential tags that define the document's framework:
r
- `<!DOCTYPE html>`: Declares the document type and tells the browser it is an HTML5 document.

- `<html>`: The root element that wraps all the content on the entire web page.

- `<head>`: Contains meta-information about the HTML page, such as the document's `<title>`, character set, and links to CSS stylesheets. This content is generally not visible on the webpage itself.

- `<body>`: Contains all the visible content of the webpage, such as headings, paragraphs, images, links, and lists.

### Why it is important

Following this standardized structure ensures that web browsers correctly interpret and render the page. It is crucial for:

- Web accessibility
- Search Engine Optimization (SEO)
- Proper CSS styling
- JavaScript interactions
- Creating a predictable and bug-free webpage structure
