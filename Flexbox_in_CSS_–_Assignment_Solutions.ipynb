# Q1. Before the main differences between the CSS flexbox layout model and the CSS box model. When would you use one over the other?

## Answer:

| CSS Box Model | CSS Flexbox Model |
|---|---|
| Used for spacing and sizing of elements. | Used for layout and alignment of elements. |
| Works on individual elements. | Works on parent-child relationships. |
| Includes margin, border, padding, and content. | Includes flex container and flex items. |
| Best for simple layouts. | Best for responsive layouts. |
| Does not easily align items dynamically. | Easily aligns items horizontally and vertically. |

---

### When to Use CSS Box Model

- For controlling:
  - Margin
  - Padding
  - Border
  - Width and height

Example:

```css
.box {
    width: 200px;
    padding: 20px;
    border: 2px solid black;
    margin: 10px;
}
```

---

### When to Use Flexbox

- For:
  - Responsive layouts
  - Centering elements
  - Aligning items
  - Navigation bars
  - Cards and grids

Example:

```css
.container {
    display: flex;
    justify-content: center;
    align-items: center;
}
```

# Q2. Explain the Working with Flexbox with the following terms:

1. justify-content  
2. align-items  
3. gap  
4. flex-direction  
5. flex-wrap  

---

## 1. justify-content

### Explanation:

Controls horizontal alignment of flex items.

### Example:

```css
.container {
    display: flex;
    justify-content: center;
}
```

### Values:

- flex-start
- center
- flex-end
- space-between
- space-around
- space-evenly

---

## 2. align-items

### Explanation:

Controls vertical alignment of flex items.

### Example:

```css
.container {
    display: flex;
    align-items: center;
}
```

### Values:

- stretch
- center
- flex-start
- flex-end

---

## 3. gap

### Explanation:

Adds spacing between flex items.

### Example:

```css
.container {
    display: flex;
    gap: 20px;
}
```

---

## 4. flex-direction

### Explanation:

Defines the direction of flex items.

### Example:

```css
.container {
    display: flex;
    flex-direction: row;
}
```

### Values:

- row
- row-reverse
- column
- column-reverse

---

## 5. flex-wrap

### Explanation:

Controls whether items wrap onto multiple lines.

### Example:

```css
.container {
    display: flex;
    flex-wrap: wrap;
}
```

### Values:

- nowrap
- wrap
- wrap-reverse

# Q3. Write a code to center a div using CSS Flexbox.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Center Div Using Flexbox</title>

    <style>

        body {

            height: 100vh;

            display: flex;

            justify-content: center;

            align-items: center;

            margin: 0;
        }

        .box {

            width: 200px;
            height: 200px;

            background-color: lime;
        }

    </style>

</head>

<body>

    <div class="box"></div>

</body>
</html>
```

---

### Explanation:

- `justify-content: center` centers horizontally.
- `align-items: center` centers vertically.
- `height: 100vh` uses full screen height.

# Q5. Build a clone of the IRCTC login page using CSS Flexbox.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>IRCTC Clone</title>

    <style>

        body {

            margin: 0;

            font-family: Arial, sans-serif;

            background-color: #f2f2f2;
        }

        .container {

            display: flex;

            justify-content: center;

            align-items: center;

            height: 100vh;
        }

        .login-box {

            display: flex;

            width: 900px;

            background-color: white;

            box-shadow: 0px 0px 10px gray;
        }

        .left {

            width: 50%;

            padding: 30px;
        }

        .right {

            width: 50%;
        }

        .right img {

            width: 100%;
            height: 100%;
        }

        input {

            width: 100%;

            padding: 10px;

            margin-bottom: 15px;

            border: 1px solid gray;

            border-radius: 5px;
        }

        button {

            width: 100%;

            padding: 12px;

            background-color: green;

            color: white;

            border: none;

            border-radius: 5px;

            cursor: pointer;
        }

    </style>

</head>

<body>

    <div class="container">

        <div class="login-box">

            <div class="left">

                <h2>Indian Railways</h2>

                <h3>Book Ticket</h3>

                <input type="text" placeholder="From">

                <input type="text" placeholder="To">

                <input type="date">

                <button>Search Train</button>

            </div>

            <div class="right">

                <img src="https://www.istockphoto.com/photos/india-train" alt="Train">

            </div>

        </div>

    </div>

</body>
</html>
```

---

### Explanation:

- Flexbox is used for layout alignment.
- `display: flex` creates flexible containers.
- `justify-content` and `align-items` align elements properly.
- `gap` creates spacing between items.
- `flex-wrap` makes layouts responsive.
