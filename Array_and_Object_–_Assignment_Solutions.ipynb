# Q1. Shopping Cart Operations

## Answer:

```javascript
let shoppingCart = ["Milk", "Coffee", "Tea", "Honey"];

// Add Meat at the beginning
shoppingCart.unshift("Meat");

// Add Sugar at the end
shoppingCart.push("Sugar");

// Remove Honey
shoppingCart.splice(shoppingCart.indexOf("Honey"), 1);

// Change Tea to Green Tea
shoppingCart[shoppingCart.indexOf("Tea")] = "Green Tea";

console.log(shoppingCart);
```

---

## Output

```text
["Meat", "Milk", "Coffee", "Green Tea", "Sugar"]
```

---

### Explanation:

- `unshift()` adds element at beginning.
- `push()` adds element at end.
- `splice()` removes elements.
- Array index updates existing values.

# Q2. Operations on Student Ages Array

## Answer:

```javascript
const ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24];

// Sort array
ages.sort((a, b) => a - b);

console.log("Sorted Ages:", ages);

// Minimum age
console.log("Minimum Age:", ages[0]);

// Maximum age
console.log("Maximum Age:", ages[ages.length - 1]);

// Median age
const median = ages[Math.floor(ages.length / 2)];

console.log("Median Age:", median);

// Average age
const sum = ages.reduce((total, age) => total + age, 0);

const average = sum / ages.length;

console.log("Average Age:", average);

// Range
const range = ages[ages.length - 1] - ages[0];

console.log("Range:", range);
```

---

## Output

```text
Sorted Ages: [19, 19, 20, 22, 24, 24, 24, 25, 25, 26]
Minimum Age: 19
Maximum Age: 26
Median Age: 24
Average Age: 22.8
Range: 7
```

---

### Explanation:

- `sort()` arranges numbers.
- `reduce()` calculates total sum.
- Average = Sum / Total Elements.

# Q3. Object Extensibility and Sealing

## Answer:

```javascript
// Student Object
const student = {

    name: "Rahul",

    age: 21
};

// a) Prevent Extensions
Object.preventExtensions(student);

// b) Check Extensible Status
const extensibleStatus = Object.isExtensible(student);

// c) Create Teacher Object
const teacher = {

    subject: "Math"
};

// d) Seal Teacher Object
Object.seal(teacher);

// e) Check Sealed Status
const sealedStatus = Object.isSealed(teacher);

// f) Print Results
console.log("Student Extensible Status:", extensibleStatus);

console.log("Teacher Sealed Status:", sealedStatus);
```

---

## Output

```text
Student Extensible Status: false
Teacher Sealed Status: true
```

---

### Explanation:

- `Object.preventExtensions()` stops adding new properties.
- `Object.isExtensible()` checks extensibility.
- `Object.seal()` prevents addition and deletion.
- `Object.isSealed()` checks seal status.

# Q4. Student Management System Using Objects

## Answer:

```javascript
const students = [

    {
        id: 1,
        firstName: "John",
        lastName: "Doe",
        age: 20,
        grade: "A"
    },

    {
        id: 2,
        firstName: "Jane",
        lastName: "Smith",
        age: 22,
        grade: "B"
    },

    {
        id: 3,
        firstName: "Bob",
        lastName: "Johnson",
        age: 21,
        grade: "A"
    }
];

// Add Student
function addStudent(student) {

    students.push(student);
}

// Update Grade
function updateGrade(id, newGrade) {

    const student = students.find(s => s.id === id);

    if(student) {

        student.grade = newGrade;
    }
}

// Delete Student
function deleteStudent(id) {

    const index = students.findIndex(s => s.id === id);

    if(index !== -1) {

        students.splice(index, 1);
    }
}

// Display Students
function displayStudents() {

    console.log(students);
}

// Average Age
function calculateAverageAge() {

    const total = students.reduce((sum, student) => sum + student.age, 0);

    return total / students.length;
}

// Testing
addStudent({

    id: 4,
    firstName: "Alice",
    lastName: "Brown",
    age: 23,
    grade: "A"
});

updateGrade(2, "A");

deleteStudent(1);

displayStudents();

console.log("Average Age:", calculateAverageAge());
```

---

## Output

```text
[
  {
    id: 2,
    firstName: "Jane",
    lastName: "Smith",
    age: 22,
    grade: "A"
  },

  {
    id: 3,
    firstName: "Bob",
    lastName: "Johnson",
    age: 21,
    grade: "A"
  },

  {
    id: 4,
    firstName: "Alice",
    lastName: "Brown",
    age: 23,
    grade: "A"
  }
]

Average Age: 22
```

---

### Explanation:

- Objects store student information.
- Functions perform add, update, delete operations.
- `reduce()` calculates average age.

# Q5. Iterate Over Object Properties Using `for...in`

## Answer:

```javascript
const student = {

    name: "Alice",

    age: 22,

    major: "Computer Science",

    GPA: 3.8,

    isEnrolled: true
};

function displayStudentInfo(student) {

    for(let property in student) {

        console.log(`Property: ${property}, Value: ${student[property]}`);
    }
}

// Function Call
displayStudentInfo(student);
```

---

## Output

```text
Property: name, Value: Alice
Property: age, Value: 22
Property: major, Value: Computer Science
Property: GPA, Value: 3.8
Property: isEnrolled, Value: true
```

---

### Explanation:

- `for...in` loop iterates through object properties.
- `student[property]` accesses values dynamically.
- Template literals format the output.
