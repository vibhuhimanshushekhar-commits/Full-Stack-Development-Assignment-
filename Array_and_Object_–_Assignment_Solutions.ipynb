# Q1. In the Following Shopping Cart, Remove and Edit Items

## Answer:

```javascript
let shoppingCart = ["Milk", "Coffee", "Tea", "Honey"];

// Add "Meat" at the beginning
shoppingCart.unshift("Meat");

// Add "Sugar" at the end
shoppingCart.push("Sugar");

// Remove "Honey"
shoppingCart.splice(shoppingCart.indexOf("Honey"), 1);

// Modify "Tea" to "Green Tea"
shoppingCart[shoppingCart.indexOf("Tea")] = "Green Tea";

console.log(shoppingCart);
```

---

## Output

```text
["Meat", "Milk", "Coffee", "Green Tea", "Sugar"]
```

---

### Explanation:

- `unshift()` adds item at beginning.
- `push()` adds item at end.
- `splice()` removes item.
- Array index is used to modify values.

# Q2. Perform Operations on Student Ages Array

## Answer:

```javascript
const ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24];

// Sort array
ages.sort((a, b) => a - b);

console.log("Sorted Ages:", ages);

// Minimum age
console.log("Minimum Age:", ages[0]);

// Maximum age
console.log("Maximum Age:", ages[ages.length - 1]);

// Median age
let median = ages[Math.floor(ages.length / 2)];

console.log("Median Age:", median);

// Average age
let sum = ages.reduce((total, age) => total + age, 0);

let average = sum / ages.length;

console.log("Average Age:", average);

// Age range
let range = ages[ages.length - 1] - ages[0];

console.log("Age Range:", range);
```

---

## Output

```text
Sorted Ages: [19, 19, 20, 22, 24, 24, 24, 25, 25, 26]
Minimum Age: 19
Maximum Age: 26
Median Age: 24
Average Age: 22.8
Age Range: 7
```

---

### Explanation:

- `sort()` arranges ages in ascending order.
- `reduce()` calculates total sum.
- Average = Total Sum / Number of Elements.

# Q3. Explain Different Methods to Access Object Properties

## Answer:

There are mainly two ways to access object properties in JavaScript:

1. Dot Notation
2. Bracket Notation

---

# 1. Dot Notation

## Example

```javascript
const person = {

    name: "Rahul",

    age: 22
};

console.log(person.name);

console.log(person.age);
```

---

# 2. Bracket Notation

## Example

```javascript
const student = {

    name: "Priya",

    course: "BCA"
};

console.log(student["name"]);

console.log(student["course"]);
```

---

### Explanation:

- Dot notation is simple and commonly used.
- Bracket notation is useful when property names are dynamic.

# Q4. Create a Student Management System Using Objects

## Answer:

```javascript
const students = [

    {
        id: 1,
        firstName: "John",
        lastName: "Doe",
        age: 20,
        grade: "A"
    },

    {
        id: 2,
        firstName: "Jane",
        lastName: "Smith",
        age: 22,
        grade: "B"
    },

    {
        id: 3,
        firstName: "Bob",
        lastName: "Johnson",
        age: 21,
        grade: "A"
    }
];

// Add Student
function addStudent(student) {

    students.push(student);
}

// Update Student Grade
function updateGrade(id, newGrade) {

    const student = students.find(s => s.id === id);

    if(student) {

        student.grade = newGrade;
    }
}

// Delete Student
function deleteStudent(id) {

    const index = students.findIndex(s => s.id === id);

    if(index !== -1) {

        students.splice(index, 1);
    }
}

// Display Students
function displayStudents() {

    console.log(students);
}

// Calculate Average Age
function averageAge() {

    let total = students.reduce((sum, student) => sum + student.age, 0);

    return total / students.length;
}

// Testing Functions
addStudent({

    id: 4,
    firstName: "Alice",
    lastName: "Brown",
    age: 23,
    grade: "A"
});

updateGrade(2, "A");

deleteStudent(1);

displayStudents();

console.log("Average Age:", averageAge());
```

---

## Output

```text
[
  {
    id: 2,
    firstName: "Jane",
    lastName: "Smith",
    age: 22,
    grade: "A"
  },

  {
    id: 3,
    firstName: "Bob",
    lastName: "Johnson",
    age: 21,
    grade: "A"
  },

  {
    id: 4,
    firstName: "Alice",
    lastName: "Brown",
    age: 23,
    grade: "A"
  }
]

Average Age: 22
```

---

### Explanation:

- Objects store student information.
- Functions manage adding, updating, deleting, and displaying students.
- `reduce()` calculates average age.

# Q5. Iterate Over Object Properties Using `for...in` Loop

## Answer:

```javascript
const student = {

    name: "John Doe",

    age: 20,

    grade: "A",

    school: "XYZ School"
};

for(let key in student) {

    console.log(key + ": " + student[key]);
}
```

---

## Output

```text
name: John Doe
age: 20
grade: A
school: XYZ School
```

---

### Explanation:

- `for...in` loop iterates through object properties.
- `student[key]` accesses values dynamically.
