# 1. Build a simple webpage that displays text as shown in the below image.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>Text Formatting</title>
</head>
<body>

    <p><b>This text will be bolded.</b></p>
    <p><i>This text will be italic.</i></p>
    <p><u>This text will be underlined</u></p>
    <p><mark>This text will be highlighted</mark></p>
    <p>This is normal text <sup>This will be super scripted</sup> This is normal again</p>
    <p>This is normal text <sub>This text will be subscripted</sub></p>
    <p>Normal Text <small>Small Text</small></p>
    <p><del>This text will be deleted</del></p>

</body>
</html>
```

# 4. Create an ordered list of HTML tags.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>HTML Tags List</title>
</head>
<body>

    <h2>Important HTML Tags</h2>

    <ol>

        <li>
            <b>&lt;html&gt;</b> :
            The root element that wraps all content on the page.
        </li>

        <li>
            <b>&lt;head&gt;</b> :
            Contains meta-information, links to CSS, and the title of the document.
        </li>

        <li>
            <b>&lt;body&gt;</b> :
            Contains the visible content of the web page.
        </li>

        <li>
            <b>&lt;a&gt;</b> :
            Defines a hyperlink used to link from one page to another.
        </li>

        <li>
            <b>&lt;img&gt;</b> :
            Used to embed an image into a web page.
        </li>

    </ol>

</body>
</html>
```

# 8. Create a table to display a conference schedule using rowspan and colspan.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>Conference Schedule</title>
</head>
<body>

    <h2>Conference Schedule</h2>

    <table border="1" cellpadding="10" cellspacing="0">

        <tr>
            <th>Time</th>
            <th>Room 1</th>
            <th>Room 2</th>
            <th>Room 3</th>
            <th>Room 4</th>
        </tr>

        <tr>
            <td>9:00 AM - 10:00 AM</td>
            <td colspan="4">Keynote</td>
        </tr>

        <tr>
            <td>10:30 AM - 11:30 AM</td>
            <td>Session A</td>
            <td>Session B</td>
            <td>Session C</td>
            <td rowspan="2">Session D</td>
        </tr>

        <tr>
            <td>11:30 AM - 12:00 PM</td>
            <td>Session E</td>
            <td colspan="2">Session F</td>
        </tr>

        <tr>
            <td>12:00 PM - 1:00 PM</td>
            <td colspan="4">Lunch Break</td>
        </tr>

        <tr>
            <td>1:00 PM - 2:00 PM</td>
            <td>Session G</td>
            <td>Session H</td>
            <td>Session I</td>
            <td>Session J</td>
        </tr>

        <tr>
            <td>2:00 PM - 3:00 PM</td>
            <td colspan="2">Session K</td>
            <td>Session L</td>
            <td>Session M</td>
        </tr>

    </table>

</body>
</html>
```
