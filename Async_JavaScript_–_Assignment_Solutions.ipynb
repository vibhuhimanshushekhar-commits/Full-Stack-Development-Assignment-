# Problem 1

## Question

Write an example to show that JavaScript is **single-threaded** by creating two competing tasks, one that blocks the event loop and another async function that waits for a promise.


## Answer

```javascript
// Blocking Function
function blockThread() {

    console.log("Blocking task started");

    const start = Date.now();

    // Block thread for 5 seconds
    while (Date.now() - start < 5000) {

    }

    console.log("Blocking task finished");
}

// Async Function
async function asyncTask() {

    console.log("Async task started");

    await new Promise(resolve => setTimeout(resolve, 2000));

    console.log("Async task completed");
}

// Execute Functions
blockThread();

asyncTask();

console.log("Program finished");
```

---

## Output

```text
Blocking task started
Blocking task finished
Async task started
Program finished
Async task completed
```

---

### Explanation

- JavaScript is **single-threaded**, meaning it executes one task at a time.
- The blocking function prevents other operations from running.
- The async function waits for a promise, but execution starts only after the blocking task finishes.

# Problem 2

## Question

JavaScript is often called **synchronous** and **single-threaded**, yet it handles asynchronous tasks like AJAX requests, timers, and event listeners.

### a. Explain why JavaScript does not execute asynchronously by default.

### b. Write a code snippet to prove that JavaScript is inherently synchronous.


## Answer

### (a) Explanation

JavaScript executes code **synchronously by default**, meaning statements run one after another in sequence.

Asynchronous behavior is introduced only for operations like:

- `setTimeout()`
- API requests (`fetch`, AJAX)
- Event listeners
- Promises

These tasks are managed outside the main execution thread and later handled by the **event loop**.

### (b) Code Example

```javascript
console.log("Task 1");

console.log("Task 2");

console.log("Task 3");
```

---

## Output

```text
Task 1
Task 2
Task 3
```

---

### Explanation

- JavaScript executes statements line by line.
- One task completes before the next starts.
- This proves JavaScript is synchronous by default.

# Problem 3

## Question

Modify the delay function to **chain multiple promises** so that three messages are logged **in sequence with delays**.


## Answer

```javascript
// Delay Function
function delay(message, time) {

    return new Promise(resolve => {

        setTimeout(() => {

            console.log(message);

            resolve();

        }, time);
    });
}

// Chain Promises
delay("Message 1", 1000)

.then(() => delay("Message 2", 2000))

.then(() => delay("Message 3", 3000))

.then(() => console.log("All messages displayed"));
```

---

## Output

```text
(after 1 second)
Message 1

(after 2 seconds)
Message 2

(after 3 seconds)
Message 3

All messages displayed
```

---

### Explanation

- `setTimeout()` creates delays.
- Promises are chained using `.then()`.
- Each message appears only after the previous promise resolves.

# Problem 4

## Question

What are the **different states of a Promise**, and how do they transition?


## Answer

A Promise has **three states**:

1. **Pending** → Initial state, operation is still running.

2. **Fulfilled (Resolved)** → Operation completed successfully.

3. **Rejected** → Operation failed.

### Example

```javascript
const myPromise = new Promise((resolve, reject) => {

    let success = true;

    if (success) {

        resolve("Promise resolved successfully");
    }

    else {

        reject("Promise rejected");
    }
});

myPromise

.then(result => console.log(result))

.catch(error => console.log(error));
```

---

## Output

```text
Promise resolved successfully
```

---

## State Transition

```text
Pending → Fulfilled
        OR
Pending → Rejected
```

---

### Explanation

- A promise starts in the `pending` state.
- `resolve()` changes it to `fulfilled`.
- `reject()` changes it to `rejected`.
- Once settled, the state cannot change again.

# Problem 5

## Question

How does the JavaScript **event loop handle Promises differently from `setTimeout()`**?


## Answer

Promises and `setTimeout()` are handled differently inside JavaScript.

### Promises (Microtask Queue)

- Promise callbacks (`then`, `catch`, `finally`) go into the **microtask queue**.
- They execute before timers.

### setTimeout (Callback Queue / Macrotask Queue)

- `setTimeout()` callbacks go into the **callback queue**.
- They run after the microtask queue becomes empty.

### Example

```javascript
console.log("Start");

setTimeout(() => {

    console.log("setTimeout executed");

}, 0);

Promise.resolve().then(() => {

    console.log("Promise executed");
});

console.log("End");
```

---

## Output

```text
Start
End
Promise executed
setTimeout executed
```

---

### Explanation

- Synchronous code runs first.
- Promise callbacks run next because microtasks have higher priority.
- `setTimeout()` executes later as a macrotask.
