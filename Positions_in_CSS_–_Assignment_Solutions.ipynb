# Task 1: Position Boxes Using CSS

## Problem Statement:

Write code to position 5 equal-sized (50 x 50px) boxes A, B, C, D, and E as follows:

1. Box A: 200px from the left and 200px from the top of the viewport.
2. Box B: 30px left and -30px above from the center of box A.
3. Box C: 30px right and -30px above from the center of box A.
4. Box D: 30px left and -30px below from the center of box A.
5. Box E: 30px right and -30px below from the center of box A.


## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>CSS Position Task</title>

    <style>

        .container {

            width: 50px;
            height: 50px;

            background-color: lightgreen;

            position: relative;

            margin-top: 200px;
            margin-left: 200px;

            border: 2px solid black;
        }

        .box {

            width: 50px;
            height: 50px;

            background-color: white;

            border: 1px solid black;

            text-align: center;
            line-height: 50px;

            font-weight: bold;

            position: absolute;
        }

        /* Center Box */

        .a {
            top: 0;
            left: 0;
            background-color: lightgreen;
        }

        /* Top Left */

        .b {
            top: -30px;
            left: -30px;
        }

        /* Top Right */

        .c {
            top: -30px;
            right: -30px;
        }

        /* Bottom Left */

        .d {
            bottom: -30px;
            left: -30px;
        }

        /* Bottom Right */

        .e {
            bottom: -30px;
            right: -30px;
        }

    </style>

</head>

<body>

    <div class="container">

        <div class="box a">A</div>

        <div class="box b">B</div>

        <div class="box c">C</div>

        <div class="box d">D</div>

        <div class="box e">E</div>

    </div>

</body>
</html>
```

---

### Explanation:

- `position: relative` is applied to the parent container.
- `position: absolute` is applied to child boxes.
- `top`, `bottom`, `left`, and `right` properties position the boxes around A.

# Task 2: Explain Absolute and Relative Positioning

## Answer:

| Relative Position | Absolute Position |
|---|---|
| Element moves relative to its normal position. | Element moves relative to nearest positioned parent. |
| Space remains reserved in layout. | Removed from normal document flow. |
| Uses `position: relative;` | Uses `position: absolute;` |
| Used for parent containers. | Used for exact positioning. |

---

## Example of Relative Position

```css
.box {
    position: relative;
    top: 20px;
    left: 20px;
}
```

---

## Example of Absolute Position

```css
.box {
    position: absolute;
    top: 20px;
    left: 20px;
}
```

---

### Explanation:

- Relative positioning shifts the element from its original position.
- Absolute positioning places the element relative to its nearest positioned ancestor.

# Task 3: Create Layout Using CSS Float

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Float Layout</title>

    <style>

        .container {

            width: 800px;

            margin: auto;

            border: 2px solid black;

            padding: 15px;

            overflow: auto;
        }

        .text-section {

            width: 60%;

            float: left;
        }

        .image-section {

            width: 35%;

            float: right;
        }

        .image-section img {

            width: 100%;
        }

    </style>

</head>

<body>

    <div class="container">

        <div class="text-section">

            <h2>The Earth!!!</h2>

            <p>
                Nature is Earth's masterpiece, a symphony of life composed through
                millennia. Its landscapes whisper tales of ancient mountains,
                winding rivers, and blooming forests.
            </p>

        </div>

        <div class="image-section">

            <img src="earth.jpg" alt="Earth">

        </div>

    </div>

</body>
</html>
```

---

### Explanation:

- `float: left` places text on the left side.
- `float: right` places image on the right side.
- `overflow: auto` clears floated elements automatically.

# Task 4: Create a Sticky Header

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Sticky Header</title>

    <style>

        body {
            margin: 0;
            font-family: Arial, sans-serif;
        }

        header {

            background-color: black;

            padding: 15px;

            position: sticky;

            top: 0;

            z-index: 1000;
        }

        nav a {

            color: white;

            text-decoration: none;

            margin-right: 20px;

            font-weight: bold;
        }

        .content {

            padding: 20px;
        }

    </style>

</head>

<body>

    <header>

        <nav>

            <a href="#">Home</a>

            <a href="#">Pricing</a>

            <a href="#">Contact</a>

            <a href="#">Terms and Conditions</a>

        </nav>

    </header>

    <div class="content">

        <h2>Heading 1</h2>

        <p>
            Lorem ipsum dolor sit amet consectetur adipisicing elit.
        </p>

        <br>

        <h2>Heading 2</h2>

        <p>
            Lorem ipsum dolor sit amet consectetur adipisicing elit.
        </p>

        <br>

        <h2>Heading 3</h2>

        <p>
            Lorem ipsum dolor sit amet consectetur adipisicing elit.
        </p>

    </div>

</body>
</html>
```

---

### Explanation:

- `position: sticky` keeps the header fixed during scrolling.
- `top: 0` sticks the header to the top.
- `z-index` keeps it above other elements.

# Task 5: Explain z-index with Example

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Z-Index Example</title>

    <style>

        .box1 {

            width: 200px;
            height: 200px;

            background-color: red;

            position: absolute;

            top: 50px;
            left: 50px;

            z-index: 1;
        }

        .box2 {

            width: 200px;
            height: 200px;

            background-color: blue;

            position: absolute;

            top: 100px;
            left: 100px;

            z-index: 2;
        }

    </style>

</head>

<body>

    <div class="box1"></div>

    <div class="box2"></div>

</body>
</html>
```

---

### Explanation:

- `z-index` controls stacking order.
- Higher `z-index` appears above lower `z-index`.
- Works only on positioned elements.
