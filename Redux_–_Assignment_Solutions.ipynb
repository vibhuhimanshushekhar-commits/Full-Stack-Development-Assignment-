# Task 1

## Problem Statement

What is Redux and why is it used in web application development?


## Answer

### What is Redux?

**Redux** is a state management library used in JavaScript applications, especially with React, to manage and store application state in a centralized store.

Redux helps applications manage data consistently and predictably.

### Why is Redux Used?

Redux is used because it helps:

- Manage global state centrally
- Share data between components
- Avoid prop drilling
- Make state changes predictable
- Simplify debugging

### Example

Suppose an e-commerce website has:

- User login data
- Shopping cart items
- Product information

Redux stores all this data in one centralized store so every component can access it easily.

---

## Explanation

Redux improves state management by keeping data organized and accessible throughout the application.

# Task 2

## Problem Statement

Explain the core concepts of Redux.


## Answer

Redux has the following core concepts:

### 1. Store

The **store** is a centralized place where the application state is stored.

#### Example

```javascript
const store = configureStore({

    reducer: counterReducer
});
```

---

### 2. Action

An **action** is an object that describes what happened.

#### Example

```javascript
{
    type: "LOGIN"
}
```

---

### 3. Reducer

A **reducer** is a function that updates the state based on an action.

#### Example

```javascript
function reducer(state, action) {

    switch (action.type) {

        case "LOGIN":

            return {

                ...state,

                isLoggedIn: true
            };

        default:

            return state;
    }
}
```

---

### 4. Dispatch

**Dispatch** sends actions to reducers.

#### Example

```javascript
dispatch({

    type: "LOGIN"
});
```

---

### 5. Middleware

Middleware adds extra functionality such as handling asynchronous tasks.

#### Example

Redux Thunk is middleware.

---

## Explanation

Redux works by:

**Action → Dispatch → Reducer → Updated Store**

# Task 3

## Problem Statement

What is Redux Thunk and how does it facilitate asynchronous logic in Redux?


## Answer

### What is Redux Thunk?

**Redux Thunk** is middleware that allows Redux to handle asynchronous logic.

Normally, Redux reducers only handle synchronous updates.

Thunk allows actions to return a function instead of an object.

### Why Use Redux Thunk?

It helps with:

- API requests
- Delayed actions
- Async operations
- Data fetching

### Example

```javascript
function fetchUser() {

    return async function(dispatch) {

        const response =
        await fetch("/api/user");

        const data =
        await response.json();

        dispatch({

            type: "SET_USER",

            payload: data
        });
    };
}
```

---

## Explanation

- Redux Thunk handles asynchronous code.
- API call runs first.
- After completion, Redux updates state using `dispatch()`.

# Task 5

## Problem Statement

Perform a simple state update with Redux Toolkit.

Given an initial state with a user object:

```javascript
{
    name: "Guest",
    isLoggedIn: false
}
```

Write a reducer function that updates the `isLoggedIn` property to `true` when a `"LOGIN"` action is dispatched.


## Answer

```javascript
import { createSlice } from "@reduxjs/toolkit";

const initialState = {

    name: "Guest",

    isLoggedIn: false
};

const userSlice = createSlice({

    name: "user",

    initialState,

    reducers: {

        LOGIN: (state) => {

            state.isLoggedIn = true;
        }
    }
});

export const { LOGIN } = userSlice.actions;

export default userSlice.reducer;
```

---

## Example Dispatch

```javascript
dispatch(LOGIN());
```

---

## Output

### Before Login

```javascript
{
    name: "Guest",

    isLoggedIn: false
}
```

### After Login

```javascript
{
    name: "Guest",

    isLoggedIn: true
}
```

---

## Explanation

- `createSlice()` creates reducers and actions automatically.
- `LOGIN()` updates `isLoggedIn` to `true`.
- Redux Toolkit uses Immer, allowing direct state mutation safely.
