# Task 2: Write Code to Arrange Containers with Texts A, B, C, and D

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Grid Layout</title>

    <style>

        .container {

            display: grid;

            grid-template-columns: 2fr 1fr;

            grid-template-rows: 100px 100px;

            gap: 10px;

            width: 400px;

            margin: 50px auto;
        }

        .box {

            background-color: #f5deb3;

            border: 1px solid black;

            text-align: center;

            line-height: 100px;

            font-size: 24px;

            font-weight: bold;
        }

        .a {

            grid-column: 1 / 2;
        }

        .b {

            grid-column: 2 / 3;

            grid-row: 1 / 3;
        }

        .c {

            grid-column: 1 / 2;

            grid-row: 2 / 3;
        }

        .d {

            grid-column: 1 / 2;

            grid-row: 2 / 3;

            width: 50%;
            justify-self: end;
        }

    </style>

</head>

<body>

    <div class="container">

        <div class="box a">A</div>

        <div class="box b">B</div>

        <div class="box c">C</div>

        <div class="box d">D</div>

    </div>

</body>
</html>
```

---

### Explanation:

- `grid-template-columns` defines column sizes.
- `grid-template-rows` defines row sizes.
- `grid-column` controls horizontal placement.
- `grid-row` controls vertical placement.

# Task 3: Explain the Use of grid-auto-row and grid-auto-column Using Code Examples

## grid-auto-rows

### Explanation:

Defines the size of automatically created rows.

### Example:

```css
.container {

    display: grid;

    grid-template-columns: repeat(3, 1fr);

    grid-auto-rows: 100px;
}
```

### Explanation:

- Every automatically created row will have height `100px`.

---

## grid-auto-columns

### Explanation:

Defines the size of automatically created columns.

### Example:

```css
.container {

    display: grid;

    grid-auto-columns: 150px;

    grid-auto-flow: column;
}
```

### Explanation:

- Automatically created columns will have width `150px`.

# Task 4: Write CSS to Show Numbers as Shown in the Figure

## HTML

```html
<!DOCTYPE html>
<html>
<head>

    <title>Grid Number Layout</title>

    <style>

        .container {

            display: grid;

            grid-template-columns: repeat(6, 60px);

            gap: 10px;

            width: fit-content;

            margin: 50px auto;
        }

        .box {

            height: 50px;

            background-color: lightgray;

            display: flex;

            justify-content: center;

            align-items: center;

            font-size: 20px;

            font-weight: bold;
        }

        .dark {

            background-color: #333;

            color: white;
        }

    </style>

</head>

<body>

    <div class="container">

        <div class="box dark">1</div>

        <div class="box">2</div>

        <div class="box dark">3</div>

        <div class="box">4</div>

        <div class="box dark">5</div>

        <div class="box">6</div>

        <div class="box dark">7</div>

        <div class="box">8</div>

    </div>

</body>
</html>
```

---

### Explanation:

- CSS Grid creates equal-sized boxes.
- `repeat(6, 60px)` creates 6 equal columns.
- Flexbox is used inside each box for centering numbers.

# Task 5: Explain the Difference Between justify-items and justify-self Using Code Examples

## justify-items

### Explanation:

Aligns all grid items horizontally inside the grid container.

### Example:

```css
.container {

    display: grid;

    justify-items: center;
}
```

### Explanation:

- All grid items will align horizontally to the center.

---

## justify-self

### Explanation:

Aligns a single grid item horizontally.

### Example:

```css
.item {

    justify-self: end;
}
```

### Explanation:

- Only the selected item moves to the end horizontally.

---

# Difference Between justify-items and justify-self

| justify-items | justify-self |
|---|---|
| Applies to all grid items. | Applies to a single item. |
| Used on the grid container. | Used on the grid item. |
| Controls alignment of all children. | Controls alignment of one child. |
