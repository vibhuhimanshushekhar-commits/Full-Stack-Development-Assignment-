# 1 Question

Explain the purpose and importance of the OSI Model, and briefly describe each layer of the OSI model with an example.

## Answer

### What is the OSI Model?

The **OSI (Open Systems Interconnection) Model** is a conceptual framework used to understand how communication occurs between devices over a network.

It divides networking into **7 layers**, where each layer performs a specific task.

---

## Purpose of the OSI Model

- Standardizes network communication
- Helps troubleshoot networking issues
- Improves interoperability between systems
- Simplifies network design and development

---

## Importance of the OSI Model

- Makes communication structured
- Helps identify problems layer-by-layer
- Supports compatibility between devices and software

---

## Layers of OSI Model

| Layer | Function | Example |
|---|---|---|
| **7. Application Layer** | Provides network services to users | Web browser, Email |
| **6. Presentation Layer** | Data formatting, encryption, compression | SSL, JPEG, MP3 |
| **5. Session Layer** | Manages communication sessions | Login sessions |
| **4. Transport Layer** | Ensures reliable data transfer | TCP, UDP |
| **3. Network Layer** | Handles routing and IP addressing | IP Address |
| **2. Data Link Layer** | Transfers data between devices | MAC Address, Switch |
| **1. Physical Layer** | Transmits raw bits through media | Cables, Wi-Fi Signals |

---

## Example

When opening a website:

1. Application Layer → Browser sends request  
2. Transport Layer → TCP ensures delivery  
3. Network Layer → IP routes packets  
4. Physical Layer → Data travels through cables/Wi-Fi

---

## Explanation

The OSI model helps us understand how data travels from one system to another step by step.

# 2 Question

Explain the purpose and importance of TCP/IP, and briefly describe the different layers of the TCP/IP model.


## Answer

### What is TCP/IP?

**TCP/IP (Transmission Control Protocol/Internet Protocol)** is a networking model used for communication over the internet.

It defines how data is transmitted between devices.

---

## Purpose of TCP/IP

- Enables internet communication
- Transfers data reliably
- Connects devices globally

---

## Importance of TCP/IP

- Backbone of the internet
- Supports communication between different devices
- Provides reliable data transfer

---

## Layers of TCP/IP Model

| Layer | Function | Example |
|---|---|---|
| **Application Layer** | User interaction and network services | HTTP, FTP, SMTP |
| **Transport Layer** | Reliable communication | TCP, UDP |
| **Internet Layer** | Addressing and routing | IP |
| **Network Access Layer** | Physical transmission | Ethernet, Wi-Fi |

---

## Example

When visiting a website:

- Application Layer → Browser sends request using HTTP  
- Transport Layer → TCP sends packets reliably  
- Internet Layer → IP routes packets  
- Network Access Layer → Wi-Fi transmits data

---

## Explanation

TCP/IP enables devices to communicate over the internet efficiently.

# 4 Question

What is DNS, and how does it resolve domain names?


## Answer

### What is DNS?

**DNS (Domain Name System)** is called the **phonebook of the internet**.

It converts domain names into IP addresses.

Example:

```text
google.com → 142.250.183.206
```

---

## How DNS Resolves Domain Names

### Step 1

User enters:

```text
www.google.com
```

### Step 2

DNS query is sent to DNS server.

### Step 3

DNS finds the matching IP address.

### Step 4

Browser connects to the server using the IP address.

---

## Example

Instead of remembering:

```text
142.250.183.206
```

Users simply type:

```text
google.com
```

---

## Explanation

DNS makes websites easy to access using readable names.

# 5 Question

Provide a brief description of the HTTP request and response model along with an illustrative diagram.


## Answer

### What is HTTP Request–Response Model?

HTTP (**HyperText Transfer Protocol**) is a communication model between a **client** and a **server**.

The client sends a **request**, and the server sends a **response**.

---

## Steps in HTTP Request–Response

1. Client sends HTTP request  
2. Server processes request  
3. Server sends HTTP response

---

## Example

User opens:

```text
www.example.com
```

Browser sends:

```http
GET / HTTP/1.1
```

Server responds:

```http
200 OK
```

with webpage content.

---

## Illustrative Diagram

```text
+-------------+
|   Browser   |
|   (Client)  |
+-------------+
        |
        | HTTP Request
        | (GET /index.html)
        v
+-------------+
|   Web Server|
+-------------+
        |
        | HTTP Response
        | (200 OK + HTML)
        v
+-------------+
|   Browser   |
+-------------+
```

---

## Explanation

The browser requests resources, and the server returns data such as HTML, CSS, or JSON.

# 6 Question

What is CORS? Explain its purpose and why developers frequently encounter CORS errors during development.




## Answer

### What is CORS?

**CORS (Cross-Origin Resource Sharing)** is a security mechanism implemented by browsers.

It controls how resources are shared between different websites (origins).

---

## Purpose of CORS

- Prevents unauthorized access
- Protects users from malicious websites
- Allows secure cross-origin communication

---

## Example

Suppose:

Frontend:

```text
http://localhost:3000
```

Backend:

```text
http://localhost:5000
```

Since both have different origins, browsers block requests unless CORS is enabled.

---

## Why Developers Encounter CORS Errors

1. Frontend and backend run on different ports  
2. Server does not allow cross-origin requests  
3. Missing CORS headers

Example Error:

```text
Access to fetch has been blocked by CORS policy
```

---

## Solution Example (Node.js Express)

```javascript
const express = require("express");

const cors = require("cors");

const app = express();

app.use(cors());

app.listen(5000);
```

---

## Explanation

CORS protects applications by restricting unsafe cross-origin requests, but developers must configure servers properly during development.
