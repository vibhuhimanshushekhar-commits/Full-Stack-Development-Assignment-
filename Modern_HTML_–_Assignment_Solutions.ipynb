# 1. List out the features of HTML5.

## Answer:

### Features of HTML5:

1. **Semantic Elements**
   - HTML5 introduced semantic tags like `<header>`, `<footer>`, `<article>`, and `<section>` which improve webpage structure and readability.

2. **Audio and Video Support**
   - HTML5 allows embedding audio and video directly using `<audio>` and `<video>` tags without external plugins.

3. **Canvas Element**
   - The `<canvas>` tag is used for drawing graphics, animations, and games using JavaScript.

4. **Local Storage**
   - HTML5 provides local storage to store data in the browser without using cookies.

5. **Responsive Web Design Support**
   - HTML5 works well with CSS3 to create responsive websites for mobile and desktop devices.

6. **Improved Forms**
   - New input types like email, date, number, and placeholder attributes make forms more user-friendly.

# 2. What are HTML Entities? List out 5 commonly used HTML entities.

## Answer:

HTML entities are special codes used to display reserved characters or symbols in HTML.

They start with `&` and end with `;`.

### Common HTML Entities:

| Entity | Symbol | Description |
|--------|--------|-------------|
| `&lt;` | < | Less than |
| `&gt;` | > | Greater than |
| `&amp;` | & | Ampersand |
| `&copy;` | © | Copyright symbol |
| `&nbsp;` | Space | Non-breaking space |

# 3. Define accessibility in the context of web development. Discuss why it's essential to create accessible websites and how it benefits different user groups.

## Answer:

Accessibility in web development means designing websites so that everyone can use them, including people with disabilities.

Accessible websites help users with:
- Visual impairments
- Hearing impairments
- Physical disabilities
- Cognitive difficulties

### Importance of Accessibility:

1. Makes websites usable for all people.
2. Improves user experience.
3. Helps websites follow legal and ethical standards.
4. Improves SEO and search engine ranking.
5. Supports screen readers and keyboard navigation.

### Benefits for Different Users:

- Visually impaired users can use screen readers.
- Elderly users can access readable and user-friendly websites.
- Mobile users benefit from responsive and clean layouts.
- Users with slow internet can access lightweight accessible pages easily.

# 4. List any 3 ways which help us in improving the accessibility of HTML.

## Answer:

### Three ways to improve HTML accessibility:

1. **Use Semantic HTML Tags**
   - Tags like `<header>`, `<nav>`, `<main>`, and `<footer>` improve webpage structure.

2. **Add Alt Text to Images**
   - The `alt` attribute helps screen readers describe images to visually impaired users.

3. **Use Proper Headings**
   - Correct heading order (`<h1>` to `<h6>`) improves navigation and readability.

# 5. Create a web page that highlights the features of HTML5 using semantic tags.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>HTML5 Features</title>
</head>
<body>

    <header>
        <h1>Features of HTML5</h1>
    </header>

    <main>

        <section>
            <h2>Semantic Elements</h2>
            <p>
                HTML5 provides semantic tags like header, footer, article, and section for better webpage structure.
            </p>
        </section>

        <section>
            <h2>Audio and Video Support</h2>
            <p>
                HTML5 allows embedding multimedia content without external plugins.
            </p>
        </section>

        <section>
            <h2>Canvas Element</h2>
            <p>
                The canvas tag is used for graphics, animations, and games.
            </p>
        </section>

    </main>

    <footer>
        <p>Created using HTML5 semantic tags.</p>
    </footer>

</body>
</html>
```
