# 1. Write a simple program in HTML that displays the heading  
# "HTML defines the content and structure of your website” on the web browser.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>HTML Heading</title>
</head>
<body>

    <h1>HTML defines the content and structure of your website</h1>

</body>
</html>
```

# 2. Explain the purpose of comments in HTML and provide an example of how to use comments in an HTML document.

## Answer:

Comments in HTML are used to add notes or explanations inside the code. These comments are not displayed on the web page. They help developers understand the code better and make maintenance easier.

HTML comments start with `<!--` and end with `-->`.

## Example:

```html
<!DOCTYPE html>
<html>
<head>
    <title>Comments Example</title>
</head>
<body>

    <!-- This is a heading -->
    <h1>Welcome to My Website</h1>

</body>
</html>
```

# 3. Write an HTML program that includes a heading, a paragraph of text, a horizontal line, and a line break. Arrange these elements to create a simple web page layout.

## Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>Simple Web Page</title>
</head>
<body>

    <h1>Welcome to My Web Page</h1>

    <p>
        HTML is used to create the structure of web pages.
        <br>
        It is simple and easy to learn.
    </p>

    <hr>

    <p>This is another paragraph after the horizontal line.</p>

</body>
</html>
```

# 4. Write a short note on Tag and Element with an example.

## Answer:

### Tag:
A tag is a keyword enclosed within angle brackets (`< >`) that tells the browser how to display content.

Examples:
```html
<h1>, <p>, <body>
```

### Element:
An element includes the opening tag, content, and closing tag together.

Example:
```html
<p>This is a paragraph.</p>
```

Here:
- `<p>` is the opening tag
- `This is a paragraph.` is the content
- `</p>` is the closing tag

Together, they form a paragraph element.

# 5. What is the DOCTYPE Declaration in HTML?

## Answer:

The `<!DOCTYPE html>` declaration is used at the beginning of an HTML document to tell the browser that the document is written in HTML5.

It helps the browser display the webpage correctly according to HTML standards.

## Example:

```html
<!DOCTYPE html>
<html>
<head>
    <title>DOCTYPE Example</title>
</head>
<body>

    <h1>Hello World</h1>

</body>
</html>
```
