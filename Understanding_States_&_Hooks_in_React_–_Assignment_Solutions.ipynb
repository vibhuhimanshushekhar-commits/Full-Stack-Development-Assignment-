# Task 1

## Problem Statement

Discuss the concept of hooks in React. Explain how they differ from class component lifecycle methods. Also, provide an example.


## Answer

### What are Hooks in React?

Hooks are special functions introduced in React that allow functional components to use:

- State management
- Lifecycle features
- Side effects
- Context API
- References

Before hooks, these features were only available in class components.

Common React hooks include:

- `useState`
- `useEffect`
- `useContext`
- `useRef`
- `useMemo`
- `useCallback`

---

## Difference Between Hooks and Class Lifecycle Methods

| Hooks | Class Lifecycle Methods |
|---|---|
| Used in functional components | Used in class components |
| Simpler and cleaner syntax | More boilerplate code |
| Uses hooks like `useEffect()` | Uses lifecycle methods like `componentDidMount()` |
| Easier state management | Complex state handling |
| Better code reusability | Less reusable logic |

---

## Example Using Hooks

```jsx
import React, { useState } from "react";

function Counter() {

    const [count, setCount] = useState(0);

    return (

        <div>

            <h2>Count: {count}</h2>

            <button onClick={() => setCount(count + 1)}>
                Increment
            </button>

        </div>
    );
}

export default Counter;
```

---

## Explanation

- `useState(0)` creates state inside a functional component.
- `count` stores the value.
- `setCount()` updates the state.
- Clicking the button increases the count.

# Task 2

## Problem Statement

Explain and discuss its syntax, purpose, and how it manages state in functional components for the following hooks:

- `useState`
- `useEffect`

#Answer

# 1. useState Hook

## Purpose

`useState` is used to create and manage state in functional components.

---

## Syntax

```jsx
const [state, setState] = useState(initialValue);
```

---

## Example

```jsx
import React, { useState } from "react";

function Example() {

    const [name, setName] = useState("Rahul");

    return (

        <div>

            <h2>{name}</h2>

            <button onClick={() => setName("Aman")}>
                Change Name
            </button>

        </div>
    );
}

export default Example;
```

---

## Explanation

- `name` stores the current state value.
- `setName()` updates the state.
- React re-renders the component after state changes.

---

# 2. useEffect Hook

## Purpose

`useEffect` is used to handle side effects in React such as:

- API calls
- Timers
- Event listeners
- Data fetching

---

## Syntax

```jsx
useEffect(() => {

    // Side effect code

}, [dependencies]);
```

---

## Example

```jsx
import React, { useEffect } from "react";

function Example() {

    useEffect(() => {

        console.log("Component Mounted");

    }, []);

    return <h2>Hello React</h2>;
}

export default Example;
```

---

## Explanation

- `useEffect()` runs after rendering.
- Empty dependency array `[]` means it runs only once.
- Similar to `componentDidMount()` in class components.

# Task 3

## Problem Statement

Explain the concept of custom hooks in React. Provide one example scenario of when and how custom hooks can be useful, and also provide the code.


## Answer

### What are Custom Hooks?

Custom hooks are user-defined hooks that allow developers to reuse logic across multiple React components.

A custom hook is simply a JavaScript function whose name starts with `use`.

---

## Benefits of Custom Hooks

- Reusable logic
- Cleaner code
- Better maintainability
- Avoid duplicate code

---

## Example Scenario

A custom hook for tracking window width.

---

## Code Example

```jsx
import React, { useState, useEffect } from "react";

// Custom Hook
function useWindowWidth() {

    const [width, setWidth] = useState(window.innerWidth);

    useEffect(() => {

        function handleResize() {

            setWidth(window.innerWidth);
        }

        window.addEventListener("resize", handleResize);

        return () => {

            window.removeEventListener("resize", handleResize);
        };

    }, []);

    return width;
}

// Component
function App() {

    const width = useWindowWidth();

    return <h2>Window Width: {width}</h2>;
}

export default App;
```

---

## Explanation

- `useWindowWidth()` stores browser width.
- Event listener updates width dynamically.
- Logic becomes reusable in multiple components.

# Task 4

## Problem Statement

Explore ways to optimize performance in React using hooks.


## Answer

React hooks can improve performance by preventing unnecessary rendering and calculations.

---

## 1. useMemo

Used to memoize expensive calculations.

### Example

```jsx
const memoizedValue = useMemo(() => {

    return expensiveFunction(data);

}, [data]);
```

### Benefit

- Prevents recalculating values unnecessarily.

---

## 2. useCallback

Used to memoize functions.

### Example

```jsx
const handleClick = useCallback(() => {

    console.log("Button clicked");

}, []);
```

### Benefit

- Prevents unnecessary function recreation.

---

## 3. React.memo

Prevents unnecessary component re-rendering.

### Example

```jsx
const MyComponent = React.memo(function MyComponent(props) {

    return <h2>{props.name}</h2>;
});
```

### Benefit

- Component re-renders only when props change.

---

## 4. Lazy Loading

Loads components only when needed.

### Example

```jsx
const About = React.lazy(() => import("./About"));
```

### Benefit

- Improves application loading speed.

---

## Explanation

Performance optimization improves speed, reduces rendering time, and enhances user experience.

# Task 5

## Problem Statement

Implement a global state management system using the Context API and `useContext` hook. Create providers and consumers for different contexts.


## Answer

## Create Context

```jsx
import React, { createContext, useState } from "react";

// Create Context
export const UserContext = createContext();

// Provider Component
export function UserProvider({ children }) {

    const [user, setUser] = useState("Rahul");

    return (

        <UserContext.Provider value={{ user, setUser }}>

            {children}

        </UserContext.Provider>
    );
}
```

---

## Consumer Component

```jsx
import React, { useContext } from "react";

import { UserContext } from "./UserContext";

function Profile() {

    const { user } = useContext(UserContext);

    return <h2>Welcome {user}</h2>;
}

export default Profile;
```

---

## App Component

```jsx
import React from "react";

import { UserProvider } from "./UserContext";

import Profile from "./Profile";

function App() {

    return (

        <UserProvider>

            <Profile />

        </UserProvider>
    );
}

export default App;
```

---

## Explanation

- `createContext()` creates global context.
- `Provider` shares data globally.
- `useContext()` accesses shared state.
- Avoids prop drilling between components.
