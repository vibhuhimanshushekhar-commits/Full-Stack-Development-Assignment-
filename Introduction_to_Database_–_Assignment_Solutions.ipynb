# Problem 1

## Question

What is a database? Explain with an example why we need a database.


## Answer

A **database** is an organized collection of data that is stored electronically and can be easily accessed, managed, and updated.

Databases help store large amounts of information in a structured manner.

### Example

A **school management system** stores:

- Student names
- Roll numbers
- Marks
- Attendance records

Instead of storing student details in notebooks or multiple files, a database stores everything in one place and allows quick access.

### Why Do We Need a Database?

- To store data systematically
- To retrieve information quickly
- To reduce data redundancy
- To improve security
- To manage large amounts of data efficiently

---

## Explanation

A database makes data management easier and avoids confusion caused by manual storage systems.

# Problem 2

## Question

Write a short note on a file-based storage system. Explain the major challenges of a file-based storage system.


## Answer

A **file-based storage system** is a traditional method of storing data in files on a computer.

In this system, data is stored separately in text files, spreadsheets, or other formats.

### Example

A company stores:

- Employee details in one file
- Salary details in another file
- Attendance records in a separate file

### Challenges of File-Based Storage System

1. **Data Redundancy**  
   Same data may be repeated in multiple files.

2. **Data Inconsistency**  
   Duplicate data may have different values.

3. **Poor Security**  
   Data protection is difficult.

4. **Difficult Data Access**  
   Searching and retrieving data takes more time.

5. **No Proper Backup and Recovery**  
   Data loss risk is higher.

---

## Explanation

File-based systems become inefficient when data grows large and complex.

# Problem 3

## Question

What is DBMS? What was the need for DBMS?


## Answer

**DBMS (Database Management System)** is software used to create, manage, store, and retrieve databases.

It acts as an interface between users and databases.

### Examples of DBMS

- MySQL
- Oracle
- PostgreSQL
- MongoDB
- Microsoft SQL Server

### Need for DBMS

DBMS was needed because file-based systems had many problems such as:

- Data redundancy
- Data inconsistency
- Poor security
- Difficulty in accessing data
- Lack of backup and recovery

DBMS solves these issues by organizing data efficiently.

---

## Explanation

A DBMS helps manage data in a secure, fast, and organized way.

# Problem 5

## Question

List out the different types of classification in DBMS and explain.


## Answer

DBMS can be classified into the following types:

### 1. Hierarchical DBMS

Data is organized in a tree-like structure where one parent can have multiple children.

### Example

Company → Department → Employee

### 2. Network DBMS

Data is connected using many-to-many relationships.

### Example

A student can enroll in many courses.

### 3. Relational DBMS (RDBMS)

Data is stored in tables consisting of rows and columns.

### Example

MySQL, PostgreSQL

### 4. Object-Oriented DBMS (OODBMS)

Stores data as objects similar to object-oriented programming.

### Example

ObjectDB

---

## Explanation

Different DBMS types are used depending on how data relationships are managed.

# Problem 6

## Question

What is the significance of Data Modelling and explain the types of Data Modelling?

## Answer

### Significance of Data Modelling

**Data modelling** is the process of organizing and defining data structure in a database.

It helps:

- Improve database design
- Reduce redundancy
- Maintain consistency
- Improve communication between developers

### Types of Data Modelling

### 1. Conceptual Data Model

Provides a high-level view of data.

Example:

Student → Course → Teacher

### 2. Logical Data Model

Defines attributes, relationships, and constraints.

Example:

Student(ID, Name, Age)

### 3. Physical Data Model

Describes how data is stored physically in the database.

Example:

Tables, indexes, storage details

---

## Explanation

Data modelling helps in planning and organizing databases before implementation.

# Problem 7

## Question

Explain 3-schema architecture along with its advantages.


## Answer

The **3-schema architecture** is a database design model used to separate the database into different levels.

### 1. External Schema (View Level)

Defines how users view the data.

Example:

Students can only see marks and attendance.

### 2. Conceptual Schema (Logical Level)

Defines the complete structure of the database.

Example:

Tables, relationships, and constraints.

### 3. Internal Schema (Physical Level)

Defines how data is physically stored.

Example:

Storage structure and indexing.

---

## Advantages of 3-Schema Architecture

1. **Data Independence**  
   Changes at one level do not affect other levels.

2. **Better Security**  
   Different users can access different views.

3. **Easy Maintenance**  
   Database modifications become easier.

4. **Improved Data Abstraction**  
   Users only see required information.

---

## Explanation

The 3-schema architecture improves flexibility, security, and maintainability in databases.
