# Task 1

## Problem Statement

Explain the concept of portals in React and describe how they differ from regular DOM rendering.


## Answer

### What are React Portals?

**React Portals** allow components to render outside the normal DOM hierarchy of the parent component.

Normally, React components render inside their parent DOM node, but portals allow rendering into another DOM node outside the root hierarchy.

Portals are commonly used for:

- Modal dialogs
- Tooltips
- Popups
- Dropdown overlays

### Syntax

```jsx
ReactDOM.createPortal(child, container);
```

### Example

```jsx
import ReactDOM from "react-dom";

function PortalExample() {

    return ReactDOM.createPortal(

        <h2>Hello from Portal</h2>,

        document.getElementById("portal-root")
    );
}

export default PortalExample;
```

---

## Difference Between Portals and Regular DOM Rendering

| Regular Rendering | Portal Rendering |
|---|---|
| Renders inside parent component | Renders outside parent DOM |
| Follows DOM hierarchy | Bypasses DOM hierarchy |
| Used for normal UI | Used for modals, overlays |

---

## Explanation

- Regular rendering places UI inside the parent element.
- Portals render components elsewhere in the DOM.
- Helpful for UI elements that must appear above everything.

# Task 2

## Problem Statement

Create a simple React component that uses a portal to render a modal dialog. Ensure that the modal can be opened and closed using state management.


## Answer

### App.js

```jsx
import React, { useState } from "react";

import ReactDOM from "react-dom";

function Modal({ closeModal }) {

    return ReactDOM.createPortal(

        <div style={styles.overlay}>

            <div style={styles.modal}>

                <h2>Portal Modal</h2>

                <p>This modal is rendered using Portal.</p>

                <button onClick={closeModal}>

                    Close

                </button>

            </div>

        </div>,

        document.getElementById("portal-root")
    );
}

function App() {

    const [showModal, setShowModal] = useState(false);

    return (

        <div>

            <h1>React Portal Example</h1>

            <button
                onClick={() => setShowModal(true)}
            >
                Open Modal
            </button>

            {showModal && (

                <Modal
                    closeModal={() =>
                        setShowModal(false)
                    }
                />
            )}

        </div>
    );
}

const styles = {

    overlay: {

        position: "fixed",

        top: 0,

        left: 0,

        width: "100%",

        height: "100%",

        backgroundColor:
        "rgba(0,0,0,0.5)",

        display: "flex",

        justifyContent: "center",

        alignItems: "center"
    },

    modal: {

        background: "white",

        padding: "20px",

        borderRadius: "10px"
    }
};

export default App;
```

### index.html

```html
<div id="root"></div>

<div id="portal-root"></div>
```

---

## Explanation

- `createPortal()` renders modal outside root DOM.
- `useState()` manages modal open/close state.
- Clicking button opens and closes modal.

# Task 3

## Problem Statement

What is React Fiber, and what are its primary goals? Discuss how it improves the reconciliation process compared to previous versions of React.


## Answer

### What is React Fiber?

**React Fiber** is the new reconciliation engine introduced in React 16.

It improves how React updates and renders UI efficiently.

### Primary Goals of React Fiber

1. Improve rendering performance  
2. Split rendering work into smaller tasks  
3. Prioritize important UI updates  
4. Improve responsiveness  
5. Support asynchronous rendering

### How Fiber Improves Reconciliation

Before React Fiber:

- React updated the UI synchronously.
- Large updates blocked the browser.

After React Fiber:

- Rendering work is split into smaller units.
- React pauses, resumes, or prioritizes rendering.

### Example

Suppose a large list is rendering:

Without Fiber:
- Entire UI waits until rendering completes.

With Fiber:
- High-priority tasks (button clicks, animations) are processed first.

---

## Explanation

Fiber makes React faster, smoother, and more responsive by improving reconciliation.

# Task 4

## Problem Statement

What is React Suspense? Explain its purpose in managing asynchronous operations in React applications. Write a simple example to illustrate suspense in React.


## Answer

### What is React Suspense?

**React Suspense** is a feature that allows React to wait for asynchronous operations before rendering UI.

It is mainly used for:

- Lazy loading components
- Data fetching
- Code splitting

### Purpose of Suspense

- Improves loading experience
- Displays fallback UI while loading
- Reduces unnecessary rendering delays

### Example

### App.js

```jsx
import React, { Suspense } from "react";

const Profile = React.lazy(() =>
    import("./Profile")
);

function App() {

    return (

        <div>

            <h1>React Suspense Example</h1>

            <Suspense fallback={<h2>Loading...</h2>}>

                <Profile />

            </Suspense>

        </div>
    );
}

export default App;
```

### Profile.js

```jsx
function Profile() {

    return <h2>User Profile Loaded</h2>;
}

export default Profile;
```

---

## Explanation

- `React.lazy()` loads components lazily.
- `Suspense` waits until loading finishes.
- `fallback` displays temporary UI.

# Task 5

## Problem Statement

List and explain at least five best practices for optimizing performance in React applications. Provide examples where applicable.


## Answer

### 1. Use React.memo()

Prevents unnecessary component re-rendering.

#### Example

```jsx
const Child = React.memo(function Child() {

    return <h2>Child Component</h2>;
});
```

#### Benefit

- Re-renders only when props change.

---

### 2. Use useMemo()

Memoizes expensive calculations.

#### Example

```jsx
const value = useMemo(() => {

    return expensiveCalculation();

}, []);
```

#### Benefit

- Prevents repeated calculations.

---

### 3. Use useCallback()

Memoizes functions.

#### Example

```jsx
const handleClick = useCallback(() => {

    console.log("Clicked");

}, []);
```

#### Benefit

- Prevents function recreation.

---

### 4. Lazy Loading

Load components only when required.

#### Example

```jsx
const About = React.lazy(() =>
    import("./About")
);
```

#### Benefit

- Faster loading performance.

---

### 5. Avoid Unnecessary State

Keep state minimal.

#### Example

Bad:

```jsx
const [fullName, setFullName] =
useState(firstName + lastName);
```

Better:

```jsx
const fullName =
firstName + lastName;
```

#### Benefit

- Reduces re-renders.

---

## Explanation

Following these best practices improves speed, responsiveness, and user experience in React applications.
