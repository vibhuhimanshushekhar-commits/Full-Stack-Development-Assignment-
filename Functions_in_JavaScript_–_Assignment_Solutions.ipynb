# Q1. Create an Arrow Function Called `square`

## Answer:

```javascript
const square = (num) => {

    return num * num;
};

const result = square(5);

console.log(result);
```

---

## Output

```text
25
```

---

### Explanation:

- Arrow functions provide shorter syntax for writing functions.
- `num * num` calculates the square of the number.

# Q2. Create a Function `generateGreeting`

## Answer:

```javascript
function generateGreeting(name) {

    return `Hello, ${name}! Welcome to JavaScript.`;
}

console.log(generateGreeting("Rahul"));

console.log(generateGreeting("Ankit"));

console.log(generateGreeting("Priya"));
```

---

## Output

```text
Hello, Rahul! Welcome to JavaScript.
Hello, Ankit! Welcome to JavaScript.
Hello, Priya! Welcome to JavaScript.
```

---

### Explanation:

- Function accepts a `name` parameter.
- Template literals (`${}`) insert values dynamically.

# Q3. Create an IIFE to Calculate Square of a Number

## Answer:

```javascript
(function(num) {

    console.log(num * num);

})(6);
```

---

## Output

```text
36
```

---

### Explanation:

- IIFE stands for Immediately Invoked Function Expression.
- Function runs immediately after creation.

# Q4. Create a Function `calculateTax`

## Answer:

```javascript
function calculateTax(income) {

    let tax;

    if(income <= 250000) {

        tax = 0;
    }

    else if(income <= 500000) {

        tax = income * 0.05;
    }

    else if(income <= 1000000) {

        tax = income * 0.20;
    }

    else {

        tax = income * 0.30;
    }

    return tax;
}

console.log(calculateTax(200000));

console.log(calculateTax(400000));

console.log(calculateTax(800000));

console.log(calculateTax(1500000));
```

---

## Output

```text
0
20000
160000
450000
```

---

### Explanation:

- Different tax rates apply based on income range.
- `if...else` conditions determine tax amount.

# Q5. Create a Recursive Function to Calculate Factorial

## Answer:

```javascript
function factorial(num) {

    if(num === 0 || num === 1) {

        return 1;
    }

    return num * factorial(num - 1);
}

console.log(factorial(5));

console.log(factorial(3));

console.log(factorial(0));
```

---

## Output

```text
120
6
1
```

---

### Explanation:

- Recursion means a function calling itself.
- Factorial Formula:

```text
5! = 5 × 4 × 3 × 2 × 1
```

# Q6. Create a Currying Function

## Answer:

```javascript
function curry(fn) {

    return function(a) {

        return function(b) {

            return fn(a, b);
        };
    };
}

function add(x, y) {

    return x + y;
}

const curriedAdd = curry(add);

console.log(curriedAdd(5)(10));

console.log(curriedAdd(20)(30));
```

---

## Output

```text
15
50
```

---

### Explanation:

- Currying converts a function with multiple arguments into nested functions.
- Arguments are supplied one at a time.
