# TASK 1

## Problem Statement 1: Create a Constructor for a Person

Write a JavaScript function constructor named `Person` that takes two parameters, `name` and `age`, and add a method to the prototype to display name and age. Then, create two instances of `Person` and display their names and ages.

## Answer:

```javascript
function Person(name, age) {

    this.name = name;

    this.age = age;
}

// Prototype Method
Person.prototype.displayInfo = function() {

    console.log(`Name: ${this.name}, Age: ${this.age}`);
};

// Create Instances
const person1 = new Person("Rahul", 21);

const person2 = new Person("Priya", 23);

// Display Information
person1.displayInfo();

person2.displayInfo();
```

---

## Output

```text
Name: Rahul, Age: 21
Name: Priya, Age: 23
```

---

### Explanation:

- Constructor function creates objects.
- `prototype` adds shared methods.
- `new` keyword creates instances.

## Problem Statement 2: Implement a Bank Account

Create a constructor function named `BankAccount` that initializes a bank account with an initial balance. Include methods for depositing and withdrawing money from the account. Write code to demonstrate these operations on a bank account object.

## Answer:

```javascript
function BankAccount(balance) {

    this.balance = balance;
}

// Deposit Method
BankAccount.prototype.deposit = function(amount) {

    this.balance += amount;

    console.log(`Deposited: ${amount}`);
};

// Withdraw Method
BankAccount.prototype.withdraw = function(amount) {

    if(amount <= this.balance) {

        this.balance -= amount;

        console.log(`Withdrawn: ${amount}`);
    }

    else {

        console.log("Insufficient Balance");
    }
};

// Display Balance
BankAccount.prototype.displayBalance = function() {

    console.log(`Current Balance: ${this.balance}`);
};

// Create Account
const account = new BankAccount(5000);

account.deposit(2000);

account.withdraw(1000);

account.displayBalance();
```

---

## Output

```text
Deposited: 2000
Withdrawn: 1000
Current Balance: 6000
```

---

### Explanation:

- `deposit()` adds money.
- `withdraw()` removes money.
- Balance updates dynamically.

## Problem Statement 5: Implement a Calculator Class

Create a class called `Calculator` that initializes two values `value1` and `value2` to store numbers. Add methods for add, subtract, multiply, and divide. Perform and show operations.

## Answer:

```javascript
class Calculator {

    constructor(value1, value2) {

        this.value1 = value1;

        this.value2 = value2;
    }

    add() {

        return this.value1 + this.value2;
    }

    subtract() {

        return this.value1 - this.value2;
    }

    multiply() {

        return this.value1 * this.value2;
    }

    divide() {

        return this.value1 / this.value2;
    }
}

// Create Object
const calc = new Calculator(10, 5);

// Perform Operations
console.log("Addition:", calc.add());

console.log("Subtraction:", calc.subtract());

console.log("Multiplication:", calc.multiply());

console.log("Division:", calc.divide());
```

---

## Output

```text
Addition: 15
Subtraction: 5
Multiplication: 50
Division: 2
```

---

### Explanation:

- Class stores two values.
- Methods perform arithmetic operations.

## Problem Statement 6: Design a Class for Geometric Shape

Design a class called `Shape` that can represent various geometric shapes. Implement subclasses for specific shapes like `Circle`, `Rectangle`, and `RightTriangle`. Each subclass should have properties of that shape (e.g., width, height for rectangle) and methods for calculating the area and perimeter of the shape. Create instances of these shapes and calculate their areas and perimeters.

## Answer:

```javascript
// Parent Class
class Shape {

    area() {

        return 0;
    }

    perimeter() {

        return 0;
    }
}

// Circle Class
class Circle extends Shape {

    constructor(radius) {

        super();

        this.radius = radius;
    }

    area() {

        return Math.PI * this.radius * this.radius;
    }

    perimeter() {

        return 2 * Math.PI * this.radius;
    }
}

// Rectangle Class
class Rectangle extends Shape {

    constructor(width, height) {

        super();

        this.width = width;

        this.height = height;
    }

    area() {

        return this.width * this.height;
    }

    perimeter() {

        return 2 * (this.width + this.height);
    }
}

// RightTriangle Class
class RightTriangle extends Shape {

    constructor(base, height, hypotenuse) {

        super();

        this.base = base;

        this.height = height;

        this.hypotenuse = hypotenuse;
    }

    area() {

        return 0.5 * this.base * this.height;
    }

    perimeter() {

        return this.base + this.height + this.hypotenuse;
    }
}

// Create Objects
const circle = new Circle(7);

const rectangle = new Rectangle(10, 5);

const triangle = new RightTriangle(3, 4, 5);

// Display Results
console.log("Circle Area:", circle.area());

console.log("Circle Perimeter:", circle.perimeter());

console.log("Rectangle Area:", rectangle.area());

console.log("Rectangle Perimeter:", rectangle.perimeter());

console.log("Triangle Area:", triangle.area());

console.log("Triangle Perimeter:", triangle.perimeter());
```

---

## Output

```text
Circle Area: 153.93
Circle Perimeter: 43.98
Rectangle Area: 50
Rectangle Perimeter: 30
Triangle Area: 6
Triangle Perimeter: 12
```

---

### Explanation:

- `extends` enables inheritance.
- Child classes override parent methods.
- Each shape calculates its own area and perimeter.

# TASK 7

## Problem Statement 7: Implement a Library System

Create a class called `Library` that can manage a collection of books. Each book should be represented as an instance of a `Book` class (with properties like title, author, and availability). Implement methods in the `Library` class to add books, check out books, and return books. Keep track of the books available.

## Answer:

```javascript
class Book {

    constructor(title, author, available = true) {

        this.title = title;

        this.author = author;

        this.available = available;
    }
}

class Library {

    constructor() {

        this.books = [];
    }

    // Add Book
    addBook(book) {

        this.books.push(book);

        console.log(`${book.title} added to library`);
    }

    // Check Out Book
    checkoutBook(title) {

        const book = this.books.find(book => book.title === title);

        if(book && book.available) {

            book.available = false;

            console.log(`${title} checked out`);
        }

        else {

            console.log(`${title} is not available`);
        }
    }

    // Return Book
    returnBook(title) {

        const book = this.books.find(book => book.title === title);

        if(book) {

            book.available = true;

            console.log(`${title} returned`);
        }
    }

    // Display Books
    displayBooks() {

        console.log(this.books);
    }
}

// Create Library
const library = new Library();

// Add Books
library.addBook(new Book("Harry Potter", "J.K. Rowling"));

library.addBook(new Book("The Alchemist", "Paulo Coelho"));

// Checkout Book
library.checkoutBook("Harry Potter");

// Return Book
library.returnBook("Harry Potter");

// Display Books
library.displayBooks();
```

---

## Output

```text
Harry Potter added to library
The Alchemist added to library
Harry Potter checked out
Harry Potter returned
[
  Book {
    title: 'Harry Potter',
    author: 'J.K. Rowling',
    available: true
  },
  Book {
    title: 'The Alchemist',
    author: 'Paulo Coelho',
    available: true
  }
]
```

---

### Explanation:

- `Library` manages book collection.
- `checkoutBook()` marks book unavailable.
- `returnBook()` marks book available again.
