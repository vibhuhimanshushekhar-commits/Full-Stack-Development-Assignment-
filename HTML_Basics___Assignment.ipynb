## Question 1: Explain the basic structure of an HTML page. Why is it important to follow this structure?

### Answer:

The basic structure of an HTML page is essentially the skeleton of a website. It starts off with `<!DOCTYPE html>` at the very top to let the browser know it's an HTML5 file. After that, everything is wrapped inside the `<html>` tags.

Inside the HTML tags, there are two main sections:

1. `<head>`: This section contains behind-the-scenes information that does not directly appear on the webpage, such as the page title, meta tags, and links to CSS files.

2. `<body>`: This section contains all the visible content shown on the webpage, including headings, paragraphs, images, buttons, and links.

It is important to follow this structure because browsers use it to correctly understand and display the webpage. If the structure is not followed properly, the webpage may not render correctly, and accessibility tools or search engines may also face issues understanding the content.

### Example:

```html
<!DOCTYPE html>
<html>
<head>
    <title>My Page</title>
</head>
<body>
    <h1>Hello World</h1>
</body>
</html>
```

## Question 2: You're building a small personal profile page. Which HTML tags will you use to highlight your name as a heading, describe yourself in a paragraph, and emphasize a key skill?

### Answer:

- To display my name as a heading, I would use the `<h1>` tag because it represents the main heading of the page.

- To describe myself, I would use the `<p>` tag for writing a paragraph.

- To highlight an important skill, I would use the `<strong>` tag to make the text bold or the `<em>` tag to italicize it for emphasis.

### Example:

```html
<h1>Himanshu Shekhar</h1>

<p>I am a web development student learning HTML.</p>

<p>My key skill is <strong>Problem Solving</strong>.</p>
```

## Question 4: You're creating a form that asks the user to enter their name, select their country from a list, and click a button to submit. Write the HTML code for this.

### Answer:

```html
<!DOCTYPE html>
<html>
<head>
    <title>User Form</title>
</head>
<body>

    <h2>User Information Form</h2>

    <form>

        <label for="name">Name:</label>
        <input type="text" id="name" name="name">

        <br><br>

        <label for="country">Country:</label>

        <select id="country" name="country">
            <option value="india">India</option>
            <option value="usa">USA</option>
            <option value="uk">UK</option>
            <option value="other">Other</option>
        </select>

        <br><br>

        <button type="submit">Submit</button>

    </form>

</body>
</html>
```
