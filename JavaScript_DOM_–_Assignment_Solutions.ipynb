# Problem 1

## Question

Explain the DOM and its role in Web development.


## Answer

The **DOM (Document Object Model)** is a programming interface for web documents. It represents an HTML document as a **tree structure**, where every element, attribute, and text becomes an object.

JavaScript uses the DOM to:

- Access HTML elements
- Modify content dynamically
- Change styles
- Add or remove elements
- Handle user interactions

### Example

```javascript
document.getElementById("heading").innerText = "Hello DOM";
```

---

## Explanation

- `document` represents the webpage.
- `getElementById()` selects an HTML element.
- `innerText` changes the text content dynamically.

# Problem 2

## Question

Explain the concept of **event delegation** and provide a scenario where it is beneficial.


## Answer

**Event delegation** is a technique in JavaScript where a parent element handles events for its child elements using **event bubbling**.

Instead of attaching event listeners to every child element, a single event listener is added to the parent.

### Example

```html
<ul id="list">
    <li>Item 1</li>
    <li>Item 2</li>
    <li>Item 3</li>
</ul>
```

```javascript
document.getElementById("list").addEventListener("click", function(event) {

    console.log(event.target.innerText);
});
```

---

## Output

```text
Item 1
Item 2
Item 3
```

---

## Explanation

- The click event is attached to the parent `<ul>`.
- `event.target` identifies the clicked child element.
- Useful when items are dynamically added.

### Beneficial Scenario

A shopping cart where products are added dynamically. Instead of attaching click events to each product button, one event listener is attached to the parent container.

# Problem 3

## Question

Explain the concept of **Event Bubbling** in the DOM.


## Answer

**Event bubbling** is the process where an event starts from the target element and propagates upward to its parent elements.

### Example

```html
<div id="parent">

    <button id="child">Click Me</button>

</div>
```

```javascript
document.getElementById("parent").addEventListener("click", function() {

    console.log("Parent clicked");
});

document.getElementById("child").addEventListener("click", function() {

    console.log("Button clicked");
});
```

---

## Output

```text
Button clicked
Parent clicked
```

---

## Explanation

- First, the button event executes.
- Then the event bubbles to the parent `<div>`.
- Events move upward through ancestor elements.

# Problem 5

## Question

Create an HTML page with buttons to increase and decrease a counter value using JavaScript.


## Answer

```html
<!DOCTYPE html>
<html>

<head>

    <title>Counter App</title>

</head>

<body>

    <h2 id="count">0</h2>

    <button id="increase">Increase</button>

    <button id="decrease">Decrease</button>

    <script>

        let count = 0;

        const countText = document.getElementById("count");

        document.getElementById("increase")
        .addEventListener("click", function() {

            count++;

            countText.innerText = count;
        });

        document.getElementById("decrease")
        .addEventListener("click", function() {

            count--;

            countText.innerText = count;
        });

    </script>

</body>

</html>
```

---

## Output

```text
0 → Increase → 1
1 → Decrease → 0
```

---

## Explanation

- Clicking **Increase** increments the counter.
- Clicking **Decrease** decrements the counter.
- DOM updates instantly using `innerText`.

# Problem 6

## Question

Create a simple image carousel using HTML and JavaScript. Design a basic HTML structure with images, and use JavaScript to implement functionality that allows users to navigate through the images.


## Answer

```html
<!DOCTYPE html>
<html>

<head>

    <title>Image Carousel</title>

</head>

<body>

    <img id="carousel" width="400"
    src="https://via.placeholder.com/400x200?text=Image+1">

    <br><br>

    <button onclick="previousImage()">Previous</button>

    <button onclick="nextImage()">Next</button>

    <script>

        const images = [

            "https://via.placeholder.com/400x200?text=Image+1",

            "https://via.placeholder.com/400x200?text=Image+2",

            "https://via.placeholder.com/400x200?text=Image+3"
        ];

        let index = 0;

        function nextImage() {

            index = (index + 1) % images.length;

            document.getElementById("carousel").src = images[index];
        }

        function previousImage() {

            index = (index - 1 + images.length) % images.length;

            document.getElementById("carousel").src = images[index];
        }

    </script>

</body>

</html>
```

---

## Explanation

- Images are stored inside an array.
- **Next** button shows the next image.
- **Previous** button shows the previous image.
- Carousel loops continuously.

# Problem 7

## Question

Build a dynamic dropdown menu using HTML and JavaScript. Create an HTML structure for a navigation menu with dropdowns. Use JavaScript to toggle the visibility of dropdowns when the user hovers or clicks.


## Answer

```html
<!DOCTYPE html>
<html>

<head>

    <title>Dropdown Menu</title>

    <style>

        ul {

            list-style: none;
        }

        .submenu {

            display: none;
        }

    </style>

</head>

<body>

    <ul>

        <li onclick="toggleMenu()">

            Menu

            <ul class="submenu" id="submenu">

                <li>Submenu 1</li>

                <li>Submenu 2</li>

                <li>Submenu 3</li>

            </ul>

        </li>

    </ul>

    <script>

        function toggleMenu() {

            const menu = document.getElementById("submenu");

            if (menu.style.display === "block") {

                menu.style.display = "none";
            }

            else {

                menu.style.display = "block";
            }
        }

    </script>

</body>

</html>
```

---

## Explanation

- Clicking **Menu** toggles submenu visibility.
- `display: none` hides the menu initially.
- JavaScript changes display dynamically.

# Problem 8

## Question

Create a simple dynamic shopping list with the following features:

- The item should appear in the list.
- Item can be deleted using a button.
- Input should reset after entering a new item.


## Answer

```html
<!DOCTYPE html>
<html>

<head>

    <title>Shopping List</title>

</head>

<body>

    <input type="text" id="itemInput"
    placeholder="Enter item">

    <button onclick="addItem()">Add Item</button>

    <ul id="shoppingList"></ul>

    <script>

        function addItem() {

            const input =
            document.getElementById("itemInput");

            const itemText = input.value;

            if (itemText === "") {

                return;
            }

            const li =
            document.createElement("li");

            li.innerHTML = `
                ${itemText}
                <button onclick="this.parentElement.remove()">
                    Delete
                </button>
            `;

            document
            .getElementById("shoppingList")
            .appendChild(li);

            // Reset Input
            input.value = "";
        }

    </script>

</body>

</html>
```

---

## Explanation

- Items are added dynamically using `createElement()`.
- Delete button removes an item.
- Input field resets after adding an item.
