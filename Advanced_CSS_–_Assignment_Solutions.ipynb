# Task 1: Create Buttons with Normal, Hover, and Focus Effects

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Button Effects</title>

    <style>

        body {

            font-family: Arial, sans-serif;

            padding: 50px;
        }

        button {

            padding: 12px 25px;

            border: none;

            background-color: #3498db;

            color: white;

            font-size: 18px;

            cursor: pointer;

            transition: 0.3s;
        }

        button:hover {

            background-color: #2ecc71;

            transform: scale(1.1);
        }

        button:focus {

            outline: 3px solid orange;
        }

    </style>

</head>

<body>

    <button>Click Me</button>

</body>
</html>
```

---

### Explanation:

- `:hover` changes button color and size when mouse moves over it.
- `:focus` adds outline when button is selected.
- `transition` makes animation smooth.

# Task 2: Create a 3D Cube Using CSS Transform

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>3D Cube</title>

    <style>

        body {

            display: flex;

            justify-content: center;

            align-items: center;

            height: 100vh;

            perspective: 800px;

            background-color: #f2f2f2;
        }

        .cube {

            width: 150px;

            height: 150px;

            position: relative;

            transform-style: preserve-3d;

            transform: rotateX(-20deg) rotateY(30deg);

            animation: rotate 5s infinite linear;
        }

        .face {

            position: absolute;

            width: 150px;

            height: 150px;

            background: rgba(52, 152, 219, 0.8);

            border: 2px solid white;

            display: flex;

            justify-content: center;

            align-items: center;

            font-size: 25px;

            color: white;
        }

        .front  { transform: translateZ(75px); }

        .back   { transform: rotateY(180deg) translateZ(75px); }

        .right  { transform: rotateY(90deg) translateZ(75px); }

        .left   { transform: rotateY(-90deg) translateZ(75px); }

        .top    { transform: rotateX(90deg) translateZ(75px); }

        .bottom { transform: rotateX(-90deg) translateZ(75px); }

        @keyframes rotate {

            from {

                transform: rotateX(0deg) rotateY(0deg);
            }

            to {

                transform: rotateX(360deg) rotateY(360deg);
            }
        }

    </style>

</head>

<body>

    <div class="cube">

        <div class="face front">Front</div>

        <div class="face back">Back</div>

        <div class="face right">Right</div>

        <div class="face left">Left</div>

        <div class="face top">Top</div>

        <div class="face bottom">Bottom</div>

    </div>

</body>
</html>
```

---

### Explanation:

- `transform-style: preserve-3d` keeps 3D effect.
- `translateZ()` moves faces in 3D space.
- `rotateX()` and `rotateY()` rotate cube.
- `animation` continuously rotates the cube.

# Task 3: Write Code to Rotate a Loader Continuously

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Rotating Loader</title>

    <style>

        body {

            display: flex;

            justify-content: center;

            align-items: center;

            height: 100vh;
        }

        .loader {

            width: 80px;

            height: 80px;

            border: 10px solid #ddd;

            border-top: 10px solid #3498db;

            border-radius: 50%;

            animation: spin 1s linear infinite;
        }

        @keyframes spin {

            from {

                transform: rotate(0deg);
            }

            to {

                transform: rotate(360deg);
            }
        }

    </style>

</head>

<body>

    <div class="loader"></div>

</body>
</html>
```

---

### Explanation:

- `border-radius: 50%` creates circle shape.
- `animation` rotates loader infinitely.
- `linear infinite` keeps smooth continuous rotation.

# Task 4: Change the Logo by Hovering on Navigation Bar

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Logo Hover Effect</title>

    <style>

        body {

            margin: 0;

            font-family: Arial, sans-serif;
        }

        nav {

            background-color: #f5f5f5;

            padding: 20px;

            text-align: center;
        }

        .logo {

            font-size: 30px;

            font-weight: bold;

            color: black;

            transition: 0.3s;
        }

        nav:hover .logo {

            color: red;

            transform: scale(1.2);
        }

    </style>

</head>

<body>

    <nav>

        <div class="logo">PW Skills</div>

    </nav>

</body>
</html>
```

---

### Explanation:

- `nav:hover .logo` changes logo style when navbar is hovered.
- `transform: scale()` enlarges logo.
- `transition` adds smooth effect.

# Task 5: Change Text Using CSS Animation

## Answer:

```html
<!DOCTYPE html>
<html>
<head>

    <title>Changing Text Animation</title>

    <style>

        body {

            display: flex;

            justify-content: center;

            align-items: center;

            height: 100vh;

            font-family: Arial, sans-serif;
        }

        .text::after {

            content: "EASY";

            color: red;

            animation: changeText 4s infinite;
        }

        @keyframes changeText {

            0% {

                content: "EASY";
            }

            50% {

                content: "AFFORDABLE";
            }

            100% {

                content: "EASY";
            }
        }

    </style>

</head>

<body>

    <h1>Upscaling Made <span class="text"></span></h1>

</body>
</html>
```

---

### Explanation:

- `::after` inserts animated text.
- `@keyframes` changes text content automatically.
- `animation: infinite` repeats forever.
