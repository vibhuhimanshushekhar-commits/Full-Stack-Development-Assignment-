# Problem 1

## Question

Create a `Map` in JavaScript and perform the following operations:

- Add key-value pairs to the Map.
- Check if a specific key exists.
- Retrieve the value associated with a given key.
- Iterate through all key-value pairs.

---

## Answer

```javascript
// Create Map
const studentMap = new Map();

// Add key-value pairs
studentMap.set("name", "Rahul");
studentMap.set("age", 22);
studentMap.set("course", "JavaScript");

// Check if a key exists
console.log(studentMap.has("age"));

// Retrieve value using key
console.log(studentMap.get("name"));

// Iterate through all key-value pairs
for (const [key, value] of studentMap) {

    console.log(`${key}: ${value}`);
}
```

---

## Output

```text
true
Rahul

name: Rahul
age: 22
course: JavaScript
```

---

### Explanation

- `Map()` creates a new map object.
- `set()` adds key-value pairs.
- `has()` checks whether a key exists.
- `get()` retrieves the value of a key.
- `for...of` loop iterates through the map.

# Problem 2

## Question

Create a `Map` to store contact information (`name`, `age`, `email`, `location`) and implement a function to retrieve contact details by name.

---

## Answer

```javascript
// Create Contact Map
const contacts = new Map();

// Add Contact Information
contacts.set("Rahul", {
    
    age: 22,
    email: "rahul@gmail.com",
    location: "Delhi"
});

contacts.set("Priya", {

    age: 24,
    email: "priya@gmail.com",
    location: "Mumbai"
});

// Function to Retrieve Contact
function getContact(name) {

    if (contacts.has(name)) {

        console.log(contacts.get(name));
    }

    else {

        console.log("Contact not found");
    }
}

// Retrieve Contact Details
getContact("Rahul");

getContact("Priya");
```

---

## Output

```text
{
  age: 22,
  email: 'rahul@gmail.com',
  location: 'Delhi'
}

{
  age: 24,
  email: 'priya@gmail.com',
  location: 'Mumbai'
}
```

---

### Explanation

- Map stores contact details.
- Function checks if contact exists.
- `get()` retrieves the object.

# Problem 3

## Question

You are given an array of numbers named `myArray`. Create a function that takes any number of arguments and adds them to the existing array. Use the spread and rest operator.

---

## Answer

```javascript
// Existing Array
const myArray = [1, 2, 3];

// Function Using Rest Operator
function addNumbers(...numbers) {

    // Add Elements Using Spread Operator
    myArray.push(...numbers);
}

// Add Numbers
addNumbers(4, 5, 6, 7);

// Display Updated Array
console.log(myArray);
```

---

## Output

```text
[
  1, 2, 3, 4,
  5, 6, 7
]
```

---

### Explanation

- `...numbers` collects arguments into an array.
- `...numbers` inside `push()` spreads values individually.

# Problem 4

## Question

Create an object `car` with properties `brand`, `model`, and a method `displayDetails` that prints `"Brand: [brand], Model: [model]"`. Test the method using the `this` keyword.

---

## Answer

```javascript
// Create Car Object
const car = {

    brand: "Toyota",

    model: "Fortuner",

    displayDetails: function() {

        console.log(`Brand: ${this.brand}, Model: ${this.model}`);
    }
};

// Call Method
car.displayDetails();
```

---

## Output

```text
Brand: Toyota, Model: Fortuner
```

---

### Explanation

- `this.brand` refers to the current object's brand.
- `this.model` refers to the current object's model.

# Problem 5

## Question

Create two objects `person1` and `person2` with properties `name` and `age`. Create a function `introduce` that prints `"Hello, I'm [name], and I'm [age] years old."` Use the `call` method to make `person2` introduce itself using the `introduce` function.

---

## Answer

```javascript
// Create Objects
const person1 = {

    name: "Aman",

    age: 21
};

const person2 = {

    name: "Sneha",

    age: 23
};

// Create Function
function introduce() {

    console.log(`Hello, I'm ${this.name}, and I'm ${this.age} years old.`);
}

// Use call() Method
introduce.call(person1);

introduce.call(person2);
```

---

## Output

```text
Hello, I'm Aman, and I'm 21 years old.
Hello, I'm Sneha, and I'm 23 years old.
```

---

### Explanation

- `call()` invokes a function with a specific object context.
- `this.name` and `this.age` refer to the passed object.
