# Task 2

## Problem Statement

Explain the difference between **Controlled** and **Uncontrolled** components in React.

## Answer

| Controlled Component | Uncontrolled Component |
|---|---|
| Controlled by React state | Controlled by DOM |
| Uses `useState()` | Uses `useRef()` |
| React handles form data | DOM handles form data |
| Easier validation | Less control |

### Controlled Component Example

```jsx
import React, { useState } from "react";

function App() {

    const [name, setName] = useState("");

    return (

        <input
            type="text"
            value={name}
            onChange={(e) => setName(e.target.value)}
        />
    );
}
```

### Uncontrolled Component Example

```jsx
import React, { useRef } from "react";

function App() {

    const inputRef = useRef();

    function handleSubmit() {

        console.log(inputRef.current.value);
    }

    return (

        <div>

            <input
                type="text"
                ref={inputRef}
            />

            <button onClick={handleSubmit}>
                Submit
            </button>

        </div>
    );
}
```

---

## Explanation

- Controlled components store values in React state.
- Uncontrolled components access values directly using `ref`.

# Task 3

## Problem Statement

Implement form submission logic in the contact form created in Task 1. When the form is submitted, capture the form data and display it in an alert box or log it to the console.

## Answer

```jsx
import React, { useState } from "react";

function ContactForm() {

    const [formData, setFormData] = useState({

        name: "",

        email: "",

        message: ""
    });

    function handleChange(event) {

        setFormData({

            ...formData,

            [event.target.name]:
            event.target.value
        });
    }

    function handleSubmit(event) {

        event.preventDefault();

        console.log(formData);

        alert(
            `Name: ${formData.name}
Email: ${formData.email}
Message: ${formData.message}`
        );
    }

    return (

        <form onSubmit={handleSubmit}>

            <label>Name:</label>

            <br />

            <input
                type="text"
                name="name"
                onChange={handleChange}
            />

            <br /><br />

            <label>Email:</label>

            <br />

            <input
                type="email"
                name="email"
                onChange={handleChange}
            />

            <br /><br />

            <label>Message:</label>

            <br />

            <textarea
                name="message"
                onChange={handleChange}
            ></textarea>

            <br /><br />

            <button type="submit">

                Submit

            </button>

        </form>
    );
}

export default ContactForm;
```

---

## Explanation

- `useState()` stores form values.
- `handleChange()` updates input values.
- `handleSubmit()` prevents page refresh.
- Form data is shown in an alert and logged in the console.

# Task 4

## Problem Statement

Create a new form with input fields for username and password. Implement the form as an **uncontrolled component** using `ref` to access the input field values upon form submission.


## Answer

```jsx
import React, { useRef } from "react";

function LoginForm() {

    const usernameRef = useRef();

    const passwordRef = useRef();

    function handleSubmit(event) {

        event.preventDefault();

        console.log(

            "Username:",
            usernameRef.current.value
        );

        console.log(

            "Password:",
            passwordRef.current.value
        );

        alert(
            `Username: ${usernameRef.current.value}
Password: ${passwordRef.current.value}`
        );
    }

    return (

        <form onSubmit={handleSubmit}>

            <label>Username:</label>

            <br />

            <input
                type="text"
                ref={usernameRef}
            />

            <br /><br />

            <label>Password:</label>

            <br />

            <input
                type="password"
                ref={passwordRef}
            />

            <br /><br />

            <button type="submit">

                Login

            </button>

        </form>
    );
}

export default LoginForm;
```

---

## Explanation

- `useRef()` accesses form values directly.
- Form is uncontrolled because React state is not used.
- Values are displayed on submit.

# Task 5

## Problem Statement

Create a Contact form using **React Hook Form**. Apply validation rules to ensure:

- Name field is required
- Email field contains a valid email address
- Message field is at least 50 characters long
- Display error messages for validation rules


## Answer

```jsx
import React from "react";

import { useForm } from "react-hook-form";

function ContactForm() {

    const {

        register,

        handleSubmit,

        formState: { errors }

    } = useForm();

    function onSubmit(data) {

        console.log(data);

        alert("Form Submitted Successfully");
    }

    return (

        <form onSubmit={handleSubmit(onSubmit)}>

            <label>Name:</label>

            <br />

            <input
                type="text"
                {...register("name", {

                    required:
                    "Name is required"
                })}
            />

            <p>{errors.name?.message}</p>

            <br />

            <label>Email:</label>

            <br />

            <input
                type="email"
                {...register("email", {

                    required:
                    "Email is required",

                    pattern: {

                        value:
                        /^[^\s@]+@[^\s@]+\.[^\s@]+$/,

                        message:
                        "Invalid email address"
                    }
                })}
            />

            <p>{errors.email?.message}</p>

            <br />

            <label>Message:</label>

            <br />

            <textarea
                {...register("message", {

                    required:
                    "Message is required",

                    minLength: {

                        value: 50,

                        message:
                        "Message must be at least 50 characters"
                    }
                })}
            ></textarea>

            <p>{errors.message?.message}</p>

            <br />

            <button type="submit">

                Submit

            </button>

        </form>
    );
}

export default ContactForm;
```

---

## Explanation

- `react-hook-form` simplifies form handling.
- `register()` applies validation rules.
- `errors` displays validation messages.
- Form submits only when inputs are valid.
